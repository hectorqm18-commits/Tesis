```python
"""
FASE 7 - Minority Game con abstencion: variacion de la memoria m con N fijo
===========================================================================
Tesis: Complejidad e informacion en un minority game con opcion de abstencion
Base: R. Mansilla, Phys. Rev. E 62, 4553 (2000), Ec. (6)

Alfabeto: 0 = comprar, 1 = vender, 2 = abstenerse

Que hace este codigo:
  1. Para cada combinacion de m y s corre 100 partidas (las mismas 100
     semillas en todas las combinaciones), con agentes CON puntaje.
     Las cadenas se reparten entre los nucleos con joblib.
  2. A cada cadena de ganadores le calcula M(d), d = 1, ..., 100, y la
     complejidad fisica C_K(l) y c_K(l), l = 1, ..., 6.
  3. Promedia sobre las 100 cadenas y dibuja, para cada s: M(d) contra d,
     M(d) en escala log-log con su ajuste, y C_K(l). Al final, una figura de
     M(1) y C_K(6) contra m.
  4. Si GUARDAR_CSV = True, guarda el promedio y todas las cadenas de cada
     combinacion, y un CSV de resumen.

gamma = K^m / N es el parametro de control. Va en los CSV, no en los titulos.

Las funciones regla_ganador, simular_con_puntaje, informacion_mutua,
sesgo_esperado, exito_al_azar, complejidad_fisica, correr_cadena, correr_N,
resumir y guardar_Md_todas son las de la Fase 6, sin cambios. ajuste_loglog es
la de notebooks/Fase 5/Fase 5.1, sin cambios.
"""

import os
import time

import numpy as np
import matplotlib.pyplot as plt

try:
    from scipy.stats import t as t_student
except ImportError:
    t_student = None
T_95 = 1.984  # t de Student, 95%, 99 g.l. (por si no hay scipy)

try:
    from joblib import Parallel, delayed
    HAY_JOBLIB = True
except ImportError:
    HAY_JOBLIB = False


# =====================================================================
# OPCIONES
# =====================================================================
# Las graficas SIEMPRE se muestran en el notebook.
# True  = ademas se guardan como PNG en CARPETA_GRAFICAS
# False = solo se muestran, no se guarda ningun PNG
DESCARGAR_GRAFICAS = True

# True  = se guardan los CSV en CARPETA_CSV
# False = no se guarda ningun CSV
GUARDAR_CSV        = False

CARPETA_CSV      = "csv"
CARPETA_GRAFICAS = "graficas"

#   -1 -> usa TODOS los nucleos
#   -2 -> usa todos menos uno (la computadora queda mas libre mientras corre)
#    1 -> un solo nucleo
N_NUCLEOS = -1


# =====================================================================
# PARAMETROS
# =====================================================================
SEMILLA_MAESTRA = 1907443528   # genera las 100 semillas de forma reproducible
N_CADENAS = 100                # numero de cadenas (semillas distintas)

N       = 101    # numero de agentes
K       = 3      # tamano del alfabeto
T       = 60000  # ciclos registrados, desde el primero (sin burn-in)

LISTA_M = [1, 2, 3, 4, 5, 6]
LISTA_S = [2, 3]

D_MAX   = 100    # M(d) se calcula para d = 1, ..., D_MAX
L_MAX   = 6      # la complejidad se calcula para l = 1, ..., L_MAX
NIVEL   = 0.95   # nivel de confianza del intervalo

ETIQUETA = "con_puntaje"


# =====================================================================
# REGLA DE GANADOR  (igual que en la Fase 6)
# =====================================================================
def regla_ganador(conteos, rng):
    """Gana el argmin sobre las K acciones, con desempate aleatorio."""
    empatados = np.flatnonzero(conteos == conteos.min())
    return int(rng.choice(empatados))


# =====================================================================
# s ESTRATEGIAS POR AGENTE, CON PUNTAJE  (igual que en la Fase 6)
# =====================================================================
def simular_con_puntaje(N, m, s, K, T, semilla):
    """Reglas del codigo base, sin burn-in. Cada agente usa la estrategia
    con mas puntos y, al final del ciclo, todas las estrategias que
    habrian acertado al ganador reciben un punto.

    Devuelve la serie de ganadores y cuantas veces acerto cada agente.
    """
    rng = np.random.default_rng(semilla)
    N_HIST = K ** m

    # estrategias[i, j, h] = accion que la estrategia j del agente i
    #                        recomienda ante la historia con indice h
    estrategias = rng.integers(0, K, size=(N, s, N_HIST))

    # puntajes[i, j] = puntos virtuales (casos de exito) de esa estrategia
    puntajes = np.zeros((N, s))

    # memoria: los m datos que el mercado tiene "al nacer"
    historia = rng.integers(0, K, size=m)

    pesos = K ** np.arange(m - 1, -1, -1)   # convierte historia -> indice
    indices = np.arange(N)

    serie_ganadores = np.zeros(T, dtype=int)
    exitos_agente = np.zeros(N, dtype=int)

    for t in range(T):

        # 1. que situacion estan viendo los agentes
        idx = int(historia @ pesos)

        # 2. cada agente toma su estrategia con mas puntos.
        #    El ruido < 1 rompe empates al azar sin alterar el orden
        #    entre puntajes enteros distintos.
        mejor = np.argmax(puntajes + rng.random((N, s)) * 0.5, axis=1)

        # 3. consulta esa estrategia en la historia actual y actua
        acciones = estrategias[indices, mejor, idx]

        # 4. se cuenta cuantos eligieron cada opcion
        conteos = np.bincount(acciones, minlength=K)

        # 5. gana la minoria
        ganador = regla_ganador(conteos, rng)

        # 6. se premian TODAS las estrategias que hubieran acertado,
        #    se hayan jugado o no (regla del codigo base / PRE 62).
        puntajes += (estrategias[:, :, idx] == ganador)

        # 7. registrar desde el primer ciclo
        serie_ganadores[t] = ganador
        exitos_agente += (acciones == ganador)

        # 8. la memoria se desplaza: entra el ganador, sale el mas viejo
        historia = np.roll(historia, -1)
        historia[-1] = ganador

    return serie_ganadores, exitos_agente


# =====================================================================
# FUNCION DE INFORMACION MUTUA  -  Ec. (6) del PRE 62  (igual que en la Fase 6)
# =====================================================================
def informacion_mutua(serie, d, K):
    """M(d) = sum_{a,b} P_ab(d) * log2[ P_ab(d) / (P_a * P_b) ]

    P_ab(d): probabilidad de encontrar el simbolo a y, d lugares despues,
             el simbolo b. Se estima contando todos los pares
             (serie[i], serie[i+d]) de la cadena.
    P_a:     densidad del simbolo a en la cadena completa (como dice
             Mansilla: "the density of the symbol alpha").

    La suma corre sobre los K*K = 9 pares (a, b). Los pares que nunca
    aparecen (P_ab = 0) no aportan, porque 0 * log(0) = 0.
    """
    x = serie[:-d]          # simbolo alpha
    y = serie[d:]           # simbolo beta, d lugares despues
    n_pares = len(x)

    # probabilidades conjuntas P_ab(d): matriz K x K
    conjunta = np.bincount(x * K + y, minlength=K * K).reshape(K, K) / n_pares

    # densidades P_a de cada simbolo en toda la cadena
    densidad = np.bincount(serie, minlength=K) / len(serie)
    producto = np.outer(densidad, densidad)      # P_a * P_b

    M_d = 0.0
    for a in range(K):
        for b in range(K):
            if conjunta[a, b] > 0:
                M_d += conjunta[a, b] * np.log2(conjunta[a, b] / producto[a, b])
    return M_d


def sesgo_esperado(K, n_pares):
    """Valor que da M(d) en una cadena SIN correlacion, solo por ser finita.

    Es aproximadamente (K-1)^2 / (2 * n * ln 2). Solo depende de T, asi
    que es la misma linea de referencia para todos los N.
    """
    return (K - 1) ** 2 / (2 * n_pares * np.log(2))


def exito_al_azar(N, K, n_muestras=200000, semilla=0):
    """Tasa de exito que tendrian N agentes eligiendo al azar.

    El 0.265 que imprimia el Segundo paso solo vale para N = 50. Aqui se
    calcula para cada N: se reparten N agentes al azar entre las K
    acciones muchas veces y se promedia la fraccion que queda en la
    accion ganadora (la de menos agentes).
    """
    rng = np.random.default_rng(semilla)
    conteos = rng.multinomial(N, [1 / K] * K, size=n_muestras)
    return (conteos.min(axis=1) / N).mean()


# =====================================================================
# COMPLEJIDAD FISICA  (igual que en la Fase 6)
# =====================================================================
def complejidad_fisica(serie, l, K):
    """Devuelve C_K(l) y c_K(l) de una cadena de ganadores.

    Una ventana de longitud l recorre la cadena de uno en uno. Cada ventana
    se convierte en un numero (en base K) para contar cuantas veces aparece
    cada subcadena eta. Con esas frecuencias relativas p(eta):

        H(l)   = -sum p(eta) log2 p(eta)
        C_K(l) = l*log2(K) - H(l)
        c_K(l) = C_K(l) / (l*log2(K))
    """
    ventanas = np.lib.stride_tricks.sliding_window_view(serie, l)
    pesos = K ** np.arange(l - 1, -1, -1)
    codigos = ventanas @ pesos
    conteos = np.bincount(codigos, minlength=K ** l)

    p = conteos[conteos > 0] / len(codigos)
    H = -np.sum(p * np.log2(p))

    maximo = l * np.log2(K)
    C = maximo - H
    return C, C / maximo


# =====================================================================
# UNA CADENA COMPLETA  (igual que en la Fase 6)
# =====================================================================
def correr_cadena(N, m, s, K, T, semilla, d_max, l_max):
    serie, exitos = simular_con_puntaje(N, m, s, K, T, int(semilla))
    Md = np.array([informacion_mutua(serie, d, K)
                   for d in range(1, d_max + 1)])
    complejidad = np.array([complejidad_fisica(serie, l, K)
                            for l in range(1, l_max + 1)])
    return {
        "Md": Md,
        "C": complejidad[:, 0],
        "c": complejidad[:, 1],
        "frec": np.bincount(serie, minlength=K) / T,
        "tasa_exito": exitos.sum() / (N * T),
    }


def correr_N(N, m, s, K, T, semillas, d_max, l_max, n_nucleos):
    """Corre todas las cadenas de un N. El resultado no depende del numero
    de nucleos: cada cadena usa solo su propia semilla y joblib devuelve
    las cadenas en el orden en que se pidieron."""
    if HAY_JOBLIB and n_nucleos != 1:
        bloque = Parallel(n_jobs=n_nucleos)(
            delayed(correr_cadena)(N, m, s, K, T, int(sem), d_max, l_max)
            for sem in semillas)
    else:
        bloque = [correr_cadena(N, m, s, K, T, int(sem), d_max, l_max)
                  for sem in semillas]
    return {
        "Md": np.array([b["Md"] for b in bloque]),
        "C": np.array([b["C"] for b in bloque]),
        "c": np.array([b["c"] for b in bloque]),
        "frec": np.array([b["frec"] for b in bloque]),
        "tasa_exito": np.array([b["tasa_exito"] for b in bloque]),
    }


def resumir(tabla, t_crit):
    """Promedio sobre las cadenas (filas), con su intervalo de confianza."""
    media = tabla.mean(axis=0)
    desv = tabla.std(axis=0, ddof=1)
    error_est = desv / np.sqrt(tabla.shape[0])
    return {"media": media, "desv": desv, "error_est": error_est,
            "ic_inf": media - t_crit * error_est,
            "ic_sup": media + t_crit * error_est}


# =====================================================================
# AJUSTE LOG-LOG  (igual que en la Fase 5)
# =====================================================================
def ajuste_loglog(d, Md):
    """Quita los ceros, pasa a log10 y ajusta una recta.

    Devuelve un diccionario con la pendiente, la ordenada al origen,
    la R^2, los puntos usados y cuantos ceros se quitaron.
    """
    validos = Md > 0                      # UNICO filtro: fuera M(d) = 0
    x = np.log10(d[validos])
    y = np.log10(Md[validos])

    pendiente, ordenada = np.polyfit(x, y, 1)   # recta y = ordenada + pendiente x
    y_ajuste = ordenada + pendiente * x

    ss_res = np.sum((y - y_ajuste) ** 2)        # suma de cuadrados residual
    ss_tot = np.sum((y - y.mean()) ** 2)        # suma de cuadrados total
    r2 = 1 - ss_res / ss_tot

    return {
        "pendiente": pendiente,
        "ordenada": ordenada,
        "r2": r2,
        "x": x, "y": y, "y_ajuste": y_ajuste,
        "n_puntos": int(validos.sum()),
        "n_ceros": int((~validos).sum()),
    }


# =====================================================================
# GUARDAR LOS CSV
# =====================================================================
def guardar_Md_todas(ruta, tabla, semillas, encabezado):
    """Mismo formato que los CSV de las fases 2 y 3."""
    distancias = np.arange(1, tabla.shape[1] + 1)
    np.savetxt(ruta, np.column_stack([np.asarray(semillas).astype(np.int64), tabla]),
               fmt=["%d"] + ["%.8e"] * tabla.shape[1], delimiter=",",
               header=(encabezado + "\nsemilla,"
                       + ",".join(f"d{d}" for d in distancias)),
               comments="# ")


def guardar_csv_combinacion(m, s, gamma, datos, semillas, t_crit, distancias,
                            longitudes, referencia):
    parametros = (f"N={N} m={m} s={s} K={K} T={T} burn_in=0 "
                  f"gamma={gamma:.6f} {ETIQUETA}")
    base = f"{ETIQUETA}_N{N}_m{m}_s{s}_T{T}"
    descripcion = f"{s} estrategias, con puntaje"

    # --- M(d): promedio ---
    r = resumir(datos["Md"], t_crit)
    np.savetxt(os.path.join(CARPETA_CSV, f"Fase7_Md_{base}_promedio.csv"),
               np.column_stack([distancias, r["media"], r["desv"],
                                r["error_est"], r["ic_inf"], r["ic_sup"],
                                referencia,
                                np.full(len(distancias), gamma)]),
               fmt=["%d"] + ["%.8e"] * 7, delimiter=",", comments="# ",
               header=(f"FASE 7. M(d) con {descripcion}, Ec. (6) PRE 62\n"
                       f"{parametros} cadenas={N_CADENAS} "
                       f"semilla_maestra={SEMILLA_MAESTRA}\n"
                       f"intervalo de confianza {int(NIVEL*100)}% con t de "
                       f"Student (t={t_crit:.4f})\n"
                       f"d,media,desv_estandar,error_estandar,ic_inferior,"
                       f"ic_superior,sesgo_esperado,gamma"))

    # --- M(d): todas las cadenas ---
    guardar_Md_todas(
        os.path.join(CARPETA_CSV, f"Fase7_Md_{base}_todas_las_cadenas.csv"),
        datos["Md"], semillas,
        f"FASE 7. M(d) de cada cadena. Fila = una semilla, columna = d\n"
        f"{parametros}")

    # --- complejidad: promedio ---
    rC = resumir(datos["C"], t_crit)
    rc = resumir(datos["c"], t_crit)
    np.savetxt(os.path.join(CARPETA_CSV,
                            f"Fase7_complejidad_{base}_promedio.csv"),
               np.column_stack([longitudes,
                                rC["media"], rC["desv"], rC["error_est"],
                                rC["ic_inf"], rC["ic_sup"],
                                rc["media"], rc["desv"], rc["error_est"],
                                rc["ic_inf"], rc["ic_sup"],
                                np.full(len(longitudes), gamma)]),
               fmt=["%d"] + ["%.8e"] * 11, delimiter=",", comments="# ",
               header=(f"FASE 7. Complejidad fisica con {descripcion}\n"
                       f"{parametros} cadenas={N_CADENAS} "
                       f"semilla_maestra={SEMILLA_MAESTRA}\n"
                       f"intervalo de confianza {int(NIVEL*100)}% con t de "
                       f"Student (t={t_crit:.4f})\n"
                       f"l,C_media,C_desv_estandar,C_error_estandar,"
                       f"C_ic_inferior,C_ic_superior,c_media,c_desv_estandar,"
                       f"c_error_estandar,c_ic_inferior,c_ic_superior,gamma"))

    # --- complejidad: todas las cadenas ---
    columnas = (",".join(f"C_l{l}" for l in longitudes) + ","
                + ",".join(f"c_l{l}" for l in longitudes))
    np.savetxt(os.path.join(CARPETA_CSV,
                            f"Fase7_complejidad_{base}_todas_las_cadenas.csv"),
               np.column_stack([np.asarray(semillas).astype(np.int64),
                                datos["C"], datos["c"]]),
               fmt=["%d"] + ["%.8e"] * (2 * len(longitudes)), delimiter=",",
               comments="# ",
               header=(f"FASE 7. Complejidad de cada cadena. Fila = una "
                       f"semilla\n{parametros}\nsemilla,{columnas}"))


# =====================================================================
# GRAFICAS
# =====================================================================
COLORES_M = dict(zip(LISTA_M, ["#2a78d6", "#eb6834", "#1baf7a", "#eda100",
                               "#e87ba4", "#008300"]))   # uno por cada m
COLORES_S = dict(zip(LISTA_S, ["#2a78d6", "#eb6834"]))   # uno por cada s
GRIS = "#52514e"
TINTA = "#0b0b0b"
SUBTITULO = (f"N={N}, K={K}, T={T}, sin burn-in, "
             f"{N_CADENAS} semillas por cada combinación")


def terminar_figura(fig, archivo):
    if DESCARGAR_GRAFICAS:
        ruta = os.path.join(CARPETA_GRAFICAS, archivo)
        fig.savefig(ruta, dpi=200)
        print(f"  Guardado: {ruta}", flush=True)
    plt.show()
    plt.close(fig)


def graficar_por_s(s, resultados, distancias, longitudes, referencia):
    # --- Grafica 1: M(d) contra d, como en las fases 2 y 3 ---
    fig, ax = plt.subplots(figsize=(10, 5.5))
    for m in LISTA_M:
        ax.plot(distancias, resultados[(m, s)]["Md"]["media"],
                color=COLORES_M[m], lw=1.8, marker="o", ms=3, label=f"m = {m}")
    ax.plot(distancias, referencia, color=GRIS, lw=1.5, ls="--",
            label="Sesgo esperado de una cadena sin correlación")
    ax.set_xlabel("Distancia d")
    ax.set_ylabel("M(d)  [bits]")
    ax.set_title(f"Información mutua promedio — {s} estrategias, con puntaje\n"
                 + SUBTITULO, loc="left", color=TINTA)
    ax.set_xlim(0, D_MAX + 1)
    ax.set_ylim(bottom=0)
    ax.grid(axis="y", color="#e5e4df", lw=0.8)
    ax.ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
    ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.13),
              ncol=4, fontsize=9)
    fig.tight_layout()
    terminar_figura(fig, f"Fase7_Md_{ETIQUETA}_N{N}_s{s}_T{T}_promedio.png")

    # --- Grafica 1b: la misma, sin m = 1 (que llega a 0.2 bits y aplasta a las demas) ---
    fig, ax = plt.subplots(figsize=(10, 5.5))
    for m in LISTA_M:
        if m == 1:
            continue
        ax.plot(distancias, resultados[(m, s)]["Md"]["media"],
                color=COLORES_M[m], lw=1.8, marker="o", ms=3, label=f"m = {m}")
    ax.plot(distancias, referencia, color=GRIS, lw=1.5, ls="--",
            label="Sesgo esperado de una cadena sin correlación")
    ax.set_xlabel("Distancia d")
    ax.set_ylabel("M(d)  [bits]")
    ax.set_title(f"Información mutua promedio, sin m = 1 — {s} estrategias, "
                 f"con puntaje\n" + SUBTITULO, loc="left", color=TINTA)
    ax.set_xlim(0, D_MAX + 1)
    ax.set_ylim(bottom=0)
    ax.grid(axis="y", color="#e5e4df", lw=0.8)
    ax.ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
    ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.13),
              ncol=3, fontsize=9)
    fig.tight_layout()
    terminar_figura(fig, f"Fase7_Md_sin_m1_{ETIQUETA}_N{N}_s{s}_T{T}_promedio.png")

    # --- Grafica 2: log10 M(d) contra log10 d, un panel por m, con su ajuste ---
    filas = int(np.ceil(len(LISTA_M) / 3))
    fig, ejes = plt.subplots(filas, 3, figsize=(12, 3.4 * filas + 1),
                             sharex=True, sharey=True, squeeze=False)
    ejes = ejes.ravel()
    for ax in ejes[len(LISTA_M):]:
        ax.set_visible(False)
    for m, ax in zip(LISTA_M, ejes):
        res = ajuste_loglog(distancias, resultados[(m, s)]["Md"]["media"])
        print(f"  Ajuste log-log s={s} m={m}: pendiente = {res['pendiente']:.4f}  "
              f"R2 = {res['r2']:.4f}  puntos = {res['n_puntos']}", flush=True)
        ax.plot(res["x"], res["y"], "o", color=COLORES_M[LISTA_M[0]], ms=3.5,
                label=f"log$_{{10}}$ M(d), promedio de {N_CADENAS} cadenas")
        ax.plot(res["x"], res["y_ajuste"], color=TINTA, lw=2, label="Ajuste lineal")
        ax.text(0.97, 0.05, (f"pendiente = {res['pendiente']:.4f}\n"
                             f"R$^2$ = {res['r2']:.4f}"),
                transform=ax.transAxes, ha="right", va="bottom", fontsize=10,
                color=TINTA, bbox=dict(boxstyle="round,pad=0.35", fc="white",
                                       ec="#e5e4df"))
        ax.set_title(f"m = {m}", loc="left", color=TINTA)
        ax.grid(color="#e5e4df", lw=0.8)
        if ax.get_subplotspec().is_last_row():
            ax.set_xlabel("log$_{10}$ d")
        if ax.get_subplotspec().is_first_col():
            ax.set_ylabel("log$_{10}$ M(d)   [M(d) en bits]")
    fig.suptitle(f"M(d) en escala log-log — {s} estrategias, con puntaje\n"
                 + SUBTITULO, x=0.06, ha="left", color=TINTA, fontsize=12)
    manijas, etiquetas = ejes[0].get_legend_handles_labels()
    fig.legend(manijas, etiquetas, frameon=False, loc="lower center", ncol=2,
               fontsize=9)
    fig.tight_layout(rect=(0, 0.05, 1, 1))
    terminar_figura(fig, f"Fase7_Md_loglog_{ETIQUETA}_N{N}_s{s}_T{T}_promedio.png")

    # --- Grafica 3: C_K(l) promedio contra l, una curva por m ---
    fig, ax = plt.subplots(figsize=(10, 5.5))
    for m in LISTA_M:
        ax.plot(longitudes, resultados[(m, s)]["C"]["media"],
                color=COLORES_M[m], lw=1.8, marker="o", ms=5, label=f"m = {m}")
    ax.set_xlabel("Longitud de la ventana l")
    ax.set_ylabel("$C_K(l)$  [bits]")
    ax.set_title(f"Complejidad física promedio — {s} estrategias, con puntaje\n"
                 + SUBTITULO, loc="left", color=TINTA)
    ax.set_xticks(longitudes)
    ax.set_ylim(bottom=0)
    ax.grid(axis="y", color="#e5e4df", lw=0.8)
    ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.13),
              ncol=6, fontsize=9)
    fig.tight_layout()
    terminar_figura(fig, f"Fase7_CK_{ETIQUETA}_N{N}_s{s}_T{T}_promedio.png")


def graficar_contra_m(resultados):
    """M(1) y C_K(6) contra m. Tienen escalas distintas: un panel cada una."""
    fig, (ax_M, ax_C) = plt.subplots(1, 2, figsize=(11, 5))
    ms = np.array(LISTA_M)

    for s in LISTA_S:
        M1 = resultados_columna(resultados, s, "Md", 0)
        C6 = resultados_columna(resultados, s, "C", L_MAX - 1)
        for ax, v in ((ax_M, M1), (ax_C, C6)):
            inf = np.maximum(v["ic_inf"], v["media"] * 0.01)
            ax.errorbar(ms, v["media"],
                        yerr=[v["media"] - inf, v["ic_sup"] - v["media"]],
                        color=COLORES_S[s], lw=1.8, marker="o", ms=6, capsize=3,
                        label=f"s = {s}")

    ax_M.set_yscale("log")
    ax_M.set_ylabel("M(1)  [bits]")
    ax_C.set_ylabel("$C_K(6)$  [bits]")
    ax_C.set_ylim(bottom=0)
    for ax in (ax_M, ax_C):
        ax.set_xlabel("Memoria m")
        ax.set_xticks(ms)
        ax.grid(axis="y", color="#e5e4df", lw=0.8)
        ax.legend(frameon=False, fontsize=9)
    fig.suptitle("M(1) y complejidad física $C_K(6)$ contra la memoria, con "
                 f"puntaje\n{SUBTITULO}", x=0.06, ha="left", color=TINTA,
                 fontsize=12)
    fig.tight_layout()
    terminar_figura(fig, f"Fase7_M1_CK6_contra_m_{ETIQUETA}_N{N}_T{T}.png")


def resultados_columna(resultados, s, medida, columna):
    """Junta, para un s, la columna indicada de cada m."""
    return {clave: np.array([resultados[(m, s)][medida][clave][columna]
                             for m in LISTA_M])
            for clave in ("media", "ic_inf", "ic_sup")}


# =====================================================================
# PROGRAMA PRINCIPAL
# =====================================================================
if __name__ == "__main__":
    inicio = time.time()

    # --- las 100 semillas (las mismas para todas las combinaciones) ---
    semillas = np.random.SeedSequence(SEMILLA_MAESTRA).generate_state(N_CADENAS)
    distancias = np.arange(1, D_MAX + 1)
    longitudes = np.arange(1, L_MAX + 1)

    t_crit = (t_student.ppf(0.5 + NIVEL / 2, N_CADENAS - 1)
              if t_student is not None else T_95)

    if GUARDAR_CSV:
        os.makedirs(CARPETA_CSV, exist_ok=True)
    if DESCARGAR_GRAFICAS:
        os.makedirs(CARPETA_GRAFICAS, exist_ok=True)
    if not HAY_JOBLIB:
        print("AVISO: joblib no esta instalado -> se usa un solo nucleo "
              "(pip install joblib)")

    plt.rcParams.update({
        "font.size": 11,
        "axes.edgecolor": GRIS,
        "axes.labelcolor": TINTA,
        "xtick.color": GRIS,
        "ytick.color": GRIS,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })

    # --- linea de referencia: solo depende de T, igual que en las fases 2 y 3 ---
    referencia = sesgo_esperado(K, T - distancias)

    # --- todas las combinaciones de m y s ---
    azar = exito_al_azar(N, K)
    resultados = {}
    resumen = []

    for s in LISTA_S:
        for m in LISTA_M:
            gamma = K ** m / N
            t0 = time.time()
            datos = correr_N(N, m, s, K, T, semillas, D_MAX, L_MAX, N_NUCLEOS)

            rMd = resumir(datos["Md"], t_crit)
            rC = resumir(datos["C"], t_crit)
            media_c = datos["c"].mean(axis=0)
            frec = datos["frec"].mean(axis=0)
            tasa_media = datos["tasa_exito"].mean()
            media_Md = rMd["media"]
            media_C = rC["media"]
            d_max = int(distancias[np.argmax(media_Md)])
            resultados[(m, s)] = {"Md": rMd, "C": rC}

            print(f"s={s} m={m}  gamma={gamma:.5f}  "
                  f"frec(0,1,2)=({frec[0]:.3f}, {frec[1]:.3f}, {frec[2]:.3f})  "
                  f"exito={tasa_media:.4f} (azar {azar:.4f})  "
                  f"M(1)={media_Md[0]:.3e}  M_max={media_Md.max():.3e} (d={d_max})  "
                  f"C(1)={media_C[0]:.3e}  C(6)={media_C[5]:.3e}  "
                  f"[{time.time() - t0:.1f} s]", flush=True)

            if GUARDAR_CSV:
                guardar_csv_combinacion(m, s, gamma, datos, semillas, t_crit,
                                        distancias, longitudes, referencia)

            resumen.append([N, m, s, gamma, tasa_media, azar,
                            media_Md[0], media_Md[1], media_Md[2], media_Md[3],
                            media_Md.max(), d_max, *media_C, *media_c])

    # --- graficas ---
    for s in LISTA_S:
        print(f"\nGraficas de s = {s}:")
        graficar_por_s(s, resultados, distancias, longitudes, referencia)
    print("\nM(1) y C_K(6) contra m:")
    graficar_contra_m(resultados)

    # --- resumen de todas las combinaciones en un solo archivo ---
    if GUARDAR_CSV:
        nombre_resumen = os.path.join(
            CARPETA_CSV,
            f"Fase7_resumen_{ETIQUETA}_N{N}_m{LISTA_M[0]}-a-{LISTA_M[-1]}"
            f"_s{'-'.join(str(s) for s in LISTA_S)}_T{T}.csv")
        columnas_C = ",".join(f"C_l{l}" for l in longitudes)
        columnas_c = ",".join(f"c_l{l}" for l in longitudes)
        np.savetxt(nombre_resumen, np.array(resumen),
                   fmt=["%d", "%d", "%d", "%.6f", "%.6f", "%.6f",
                        "%.8e", "%.8e", "%.8e", "%.8e", "%.8e", "%d"]
                       + ["%.8e"] * (2 * L_MAX),
                   delimiter=",", comments="# ",
                   header=(f"FASE 7. Resumen, con puntaje, N={N} K={K} T={T} "
                           f"cadenas={N_CADENAS}\n"
                           f"N,m,s,gamma,tasa_exito,exito_al_azar,"
                           f"M1,M2,M3,M4,M_max,d_del_maximo,"
                           f"{columnas_C},{columnas_c}"))
        print(f"\nGuardado: {nombre_resumen}")

    print(f"\nListo. Tiempo total: {(time.time() - inicio) / 60:.1f} min")
```
