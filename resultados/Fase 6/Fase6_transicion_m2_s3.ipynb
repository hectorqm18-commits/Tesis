```python
"""
FASE 6 - Minority Game con abstencion: transicion de fase al variar N
=====================================================================
Tesis: Complejidad e informacion en un minority game con opcion de abstencion
Base: R. Mansilla, Phys. Rev. E 62, 4553 (2000), Ec. (6)

Alfabeto: 0 = comprar, 1 = vender, 2 = abstenerse

Que hace este codigo:
  1. Para cada valor de N corre 100 partidas (las mismas 100 semillas
     para todos los N), con agentes de s = 3 estrategias CON puntaje.
     Las cadenas se reparten entre los nucleos con joblib.
  2. A cada cadena de ganadores le calcula:
       - M(d), Ec. (6) del PRE 62, para d = 1, ..., 100.
       - La complejidad fisica, para l = 1, ..., 6:
         H(l)   = -sum p(eta) log2 p(eta)   (ventana movil de longitud l)
         C_K(l) = l*log2(K) - H(l)
         c_K(l) = C_K(l) / (l*log2(K))
  3. Promedia sobre las 100 cadenas para cada N.
  4. Grafica los N en grupos de hasta 4 curvas: M(d) promedio y
     C_K(l) promedio contra l.
  5. Guarda, por cada N, el promedio y todas las cadenas de M(d) y de la
     complejidad, y un CSV de resumen con todos los N.
  6. Dibuja una sola figura resumen con todos los N juntos.

Si un N ya tiene sus CSV (y REUTILIZAR_CSV = True), no se vuelve a simular:
sus promedios se leen de esos archivos.

gamma = K^m / N es el parametro de control. Va en los CSV, no en los titulos.

Las funciones regla_ganador, simular_con_puntaje, informacion_mutua,
sesgo_esperado y exito_al_azar son las de notebooks/Fase 3/Paso_3.1, sin
cambios. Lo unico nuevo es la complejidad fisica y el reparto con joblib.
"""

import glob
import os
import time

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap, LogNorm
from matplotlib.ticker import NullLocator, ScalarFormatter

try:
    from scipy.stats import t as t_student
except ImportError:
    t_student = None
T_95 = 1.984  # t de Student, 95%, 99 g.l. (por si no hay scipy)

try:
    from joblib import Parallel, delayed
    HAY_JOBLIB = True
except ImportError:
    HAY_JOBLIB = False


# =====================================================================
# OPCIONES
# =====================================================================
# Las graficas SIEMPRE se muestran en el notebook.
# True  = ademas se guardan como PNG en CARPETA_GRAFICAS
# False = solo se muestran, no se guarda ningun PNG
DESCARGAR_GRAFICAS = True

# True  = se guardan los CSV en CARPETA_CSV
# False = no se guarda ningun CSV
GUARDAR_CSV        = True

# True  = si un N ya tiene sus CSV, se leen y NO se vuelve a simular
# False = se simulan todos los N otra vez (unos 10 minutos)
REUTILIZAR_CSV     = True

CARPETA_CSV      = "csv"
CARPETA_GRAFICAS = "graficas"

#   -1 -> usa TODOS los nucleos
#   -2 -> usa todos menos uno (la computadora queda mas libre mientras corre)
#    1 -> un solo nucleo
N_NUCLEOS = -1


# =====================================================================
# PARAMETROS
# =====================================================================
SEMILLA_MAESTRA = 1907443528   # genera las 100 semillas de forma reproducible
N_CADENAS = 100                # numero de cadenas (semillas distintas)

M       = 2      # memoria (brain size)
S       = 3      # estrategias por agente
K       = 3      # tamano del alfabeto
T       = 6000   # ciclos registrados, desde el primero (sin burn-in)

D_MAX   = 100    # M(d) se calcula para d = 1, ..., D_MAX
L_MAX   = 6      # la complejidad se calcula para l = 1, ..., L_MAX
NIVEL   = 0.95   # nivel de confianza del intervalo

# >>> NUMERO DE AGENTES: esta es la linea que hay que cambiar <<<
LISTA_N = [5, 7, 8, 10, 11, 13, 14, 16, 17, 19, 20, 22, 23, 25, 26, 28, 29, 31, 32, 34, 35, 37, 38, 40, 44, 50, 64, 80, 100, 125, 160, 200, 250, 320, 400, 500, 640, 800, 1000, 1600, 2500, 4000, 5000, 6001, 6400, 6500, 6560, 6562, 6601, 6700, 6800, 7000]

# Los grupos de graficas se arman solos: hasta 4 valores de N por grafica.
N_POR_GRAFICA = 4
GRUPOS = [
    ("N" + "-".join(str(n) for n in LISTA_N[i:i + N_POR_GRAFICA]),
     LISTA_N[i:i + N_POR_GRAFICA])
    for i in range(0, len(LISTA_N), N_POR_GRAFICA)
]

ETIQUETA = "con_puntaje"
DESCRIPCION = f"{S} estrategias, con puntaje"
PARAMS_ARCHIVO = f"{ETIQUETA}_m{M}_s{S}_T{T}"


# =====================================================================
# REGLA DE GANADOR  (identica a la del codigo base)
# =====================================================================
def regla_ganador(conteos, rng):
    """Gana el argmin sobre las K acciones, con desempate aleatorio."""
    empatados = np.flatnonzero(conteos == conteos.min())
    return int(rng.choice(empatados))


# =====================================================================
# s ESTRATEGIAS POR AGENTE, CON PUNTAJE  (igual que en Paso_3.1)
# =====================================================================
def simular_con_puntaje(N, m, s, K, T, semilla):
    """Reglas del codigo base, sin burn-in. Cada agente usa la estrategia
    con mas puntos y, al final del ciclo, todas las estrategias que
    habrian acertado al ganador reciben un punto.

    Devuelve la serie de ganadores y cuantas veces acerto cada agente.
    """
    rng = np.random.default_rng(semilla)
    N_HIST = K ** m

    # estrategias[i, j, h] = accion que la estrategia j del agente i
    #                        recomienda ante la historia con indice h
    estrategias = rng.integers(0, K, size=(N, s, N_HIST))

    # puntajes[i, j] = puntos virtuales (casos de exito) de esa estrategia
    puntajes = np.zeros((N, s))

    # memoria: los m datos que el mercado tiene "al nacer"
    historia = rng.integers(0, K, size=m)

    pesos = K ** np.arange(m - 1, -1, -1)   # convierte historia -> indice
    indices = np.arange(N)

    serie_ganadores = np.zeros(T, dtype=int)
    exitos_agente = np.zeros(N, dtype=int)

    for t in range(T):

        # 1. que situacion estan viendo los agentes
        idx = int(historia @ pesos)

        # 2. cada agente toma su estrategia con mas puntos.
        #    El ruido < 1 rompe empates al azar sin alterar el orden
        #    entre puntajes enteros distintos.
        mejor = np.argmax(puntajes + rng.random((N, s)) * 0.5, axis=1)

        # 3. consulta esa estrategia en la historia actual y actua
        acciones = estrategias[indices, mejor, idx]

        # 4. se cuenta cuantos eligieron cada opcion
        conteos = np.bincount(acciones, minlength=K)

        # 5. gana la minoria
        ganador = regla_ganador(conteos, rng)

        # 6. se premian TODAS las estrategias que hubieran acertado,
        #    se hayan jugado o no (regla del codigo base / PRE 62).
        puntajes += (estrategias[:, :, idx] == ganador)

        # 7. registrar desde el primer ciclo
        serie_ganadores[t] = ganador
        exitos_agente += (acciones == ganador)

        # 8. la memoria se desplaza: entra el ganador, sale el mas viejo
        historia = np.roll(historia, -1)
        historia[-1] = ganador

    return serie_ganadores, exitos_agente


# =====================================================================
# FUNCION DE INFORMACION MUTUA  -  Ec. (6) del PRE 62  (igual que en Paso_3.1)
# =====================================================================
def informacion_mutua(serie, d, K):
    """M(d) = sum_{a,b} P_ab(d) * log2[ P_ab(d) / (P_a * P_b) ]

    P_ab(d): probabilidad de encontrar el simbolo a y, d lugares despues,
             el simbolo b. Se estima contando todos los pares
             (serie[i], serie[i+d]) de la cadena.
    P_a:     densidad del simbolo a en la cadena completa (como dice
             Mansilla: "the density of the symbol alpha").

    La suma corre sobre los K*K = 9 pares (a, b). Los pares que nunca
    aparecen (P_ab = 0) no aportan, porque 0 * log(0) = 0.
    """
    x = serie[:-d]          # simbolo alpha
    y = serie[d:]           # simbolo beta, d lugares despues
    n_pares = len(x)

    # probabilidades conjuntas P_ab(d): matriz K x K
    conjunta = np.bincount(x * K + y, minlength=K * K).reshape(K, K) / n_pares

    # densidades P_a de cada simbolo en toda la cadena
    densidad = np.bincount(serie, minlength=K) / len(serie)
    producto = np.outer(densidad, densidad)      # P_a * P_b

    M_d = 0.0
    for a in range(K):
        for b in range(K):
            if conjunta[a, b] > 0:
                M_d += conjunta[a, b] * np.log2(conjunta[a, b] / producto[a, b])
    return M_d


def sesgo_esperado(K, n_pares):
    """Valor que da M(d) en una cadena SIN correlacion, solo por ser finita.

    Es aproximadamente (K-1)^2 / (2 * n * ln 2). Solo depende de T, asi
    que es la misma linea de referencia para todos los N.
    """
    return (K - 1) ** 2 / (2 * n_pares * np.log(2))


def exito_al_azar(N, K, n_muestras=200000, semilla=0):
    """Tasa de exito que tendrian N agentes eligiendo al azar.

    El 0.265 que imprimia el Segundo paso solo vale para N = 50. Aqui se
    calcula para cada N: se reparten N agentes al azar entre las K
    acciones muchas veces y se promedia la fraccion que queda en la
    accion ganadora (la de menos agentes).
    """
    rng = np.random.default_rng(semilla)
    conteos = rng.multinomial(N, [1 / K] * K, size=n_muestras)
    return (conteos.min(axis=1) / N).mean()


# =====================================================================
# COMPLEJIDAD FISICA  (lo unico nuevo de la Fase 6)
# =====================================================================
def complejidad_fisica(serie, l, K):
    """Devuelve C_K(l) y c_K(l) de una cadena de ganadores.

    Una ventana de longitud l recorre la cadena de uno en uno. Cada ventana
    se convierte en un numero (en base K) para contar cuantas veces aparece
    cada subcadena eta. Con esas frecuencias relativas p(eta):

        H(l)   = -sum p(eta) log2 p(eta)
        C_K(l) = l*log2(K) - H(l)
        c_K(l) = C_K(l) / (l*log2(K))
    """
    ventanas = np.lib.stride_tricks.sliding_window_view(serie, l)
    pesos = K ** np.arange(l - 1, -1, -1)
    codigos = ventanas @ pesos
    conteos = np.bincount(codigos, minlength=K ** l)

    p = conteos[conteos > 0] / len(codigos)
    H = -np.sum(p * np.log2(p))

    maximo = l * np.log2(K)
    C = maximo - H
    return C, C / maximo


# =====================================================================
# UNA CADENA COMPLETA  (la tarea que se reparte entre nucleos)
# =====================================================================
def correr_cadena(N, m, s, K, T, semilla, d_max, l_max):
    serie, exitos = simular_con_puntaje(N, m, s, K, T, int(semilla))
    Md = np.array([informacion_mutua(serie, d, K)
                   for d in range(1, d_max + 1)])
    complejidad = np.array([complejidad_fisica(serie, l, K)
                            for l in range(1, l_max + 1)])
    return {
        "Md": Md,
        "C": complejidad[:, 0],
        "c": complejidad[:, 1],
        "frec": np.bincount(serie, minlength=K) / T,
        "tasa_exito": exitos.sum() / (N * T),
    }


def correr_N(N, m, s, K, T, semillas, d_max, l_max, n_nucleos):
    """Corre todas las cadenas de un N. El resultado no depende del numero
    de nucleos: cada cadena usa solo su propia semilla y joblib devuelve
    las cadenas en el orden en que se pidieron."""
    if HAY_JOBLIB and n_nucleos != 1:
        bloque = Parallel(n_jobs=n_nucleos)(
            delayed(correr_cadena)(N, m, s, K, T, int(sem), d_max, l_max)
            for sem in semillas)
    else:
        bloque = [correr_cadena(N, m, s, K, T, int(sem), d_max, l_max)
                  for sem in semillas]
    return {
        "Md": np.array([b["Md"] for b in bloque]),
        "C": np.array([b["C"] for b in bloque]),
        "c": np.array([b["c"] for b in bloque]),
        "frec": np.array([b["frec"] for b in bloque]),
        "tasa_exito": np.array([b["tasa_exito"] for b in bloque]),
    }


def resumir(tabla, t_crit):
    """Promedio sobre las cadenas (filas), con su intervalo de confianza."""
    media = tabla.mean(axis=0)
    desv = tabla.std(axis=0, ddof=1)
    error_est = desv / np.sqrt(tabla.shape[0])
    return {"media": media, "desv": desv, "error_est": error_est,
            "ic_inf": media - t_crit * error_est,
            "ic_sup": media + t_crit * error_est}


# =====================================================================
# GUARDAR LOS CSV
# =====================================================================
def guardar_Md_todas(ruta, tabla, semillas, encabezado):
    """Mismo formato que los CSV de las fases 2 y 3."""
    distancias = np.arange(1, tabla.shape[1] + 1)
    np.savetxt(ruta, np.column_stack([np.asarray(semillas).astype(np.int64), tabla]),
               fmt=["%d"] + ["%.8e"] * tabla.shape[1], delimiter=",",
               header=(encabezado + "\nsemilla,"
                       + ",".join(f"d{d}" for d in distancias)),
               comments="# ")


def guardar_csv_N(N, gamma, datos, semillas, t_crit, distancias, longitudes,
                  referencia):
    parametros = (f"N={N} m={M} s={S} K={K} T={T} burn_in=0 "
                  f"gamma={gamma:.6f} {ETIQUETA}")
    base = f"{PARAMS_ARCHIVO}_N{N}"

    # --- M(d): promedio ---
    r = resumir(datos["Md"], t_crit)
    np.savetxt(os.path.join(CARPETA_CSV, f"Fase6_Md_{base}_promedio.csv"),
               np.column_stack([distancias, r["media"], r["desv"],
                                r["error_est"], r["ic_inf"], r["ic_sup"],
                                referencia,
                                np.full(len(distancias), gamma)]),
               fmt=["%d"] + ["%.8e"] * 7, delimiter=",", comments="# ",
               header=(f"FASE 6. M(d) con {DESCRIPCION}, Ec. (6) PRE 62\n"
                       f"{parametros} cadenas={N_CADENAS} "
                       f"semilla_maestra={SEMILLA_MAESTRA}\n"
                       f"intervalo de confianza {int(NIVEL*100)}% con t de "
                       f"Student (t={t_crit:.4f})\n"
                       f"d,media,desv_estandar,error_estandar,ic_inferior,"
                       f"ic_superior,sesgo_esperado,gamma"))

    # --- M(d): todas las cadenas ---
    guardar_Md_todas(
        os.path.join(CARPETA_CSV, f"Fase6_Md_{base}_todas_las_cadenas.csv"),
        datos["Md"], semillas,
        f"FASE 6. M(d) de cada cadena. Fila = una semilla, columna = d\n"
        f"{parametros}")

    # --- complejidad: promedio ---
    rC = resumir(datos["C"], t_crit)
    rc = resumir(datos["c"], t_crit)
    np.savetxt(os.path.join(CARPETA_CSV,
                            f"Fase6_complejidad_{base}_promedio.csv"),
               np.column_stack([longitudes,
                                rC["media"], rC["desv"], rC["error_est"],
                                rC["ic_inf"], rC["ic_sup"],
                                rc["media"], rc["desv"], rc["error_est"],
                                rc["ic_inf"], rc["ic_sup"],
                                np.full(len(longitudes), gamma)]),
               fmt=["%d"] + ["%.8e"] * 11, delimiter=",", comments="# ",
               header=(f"FASE 6. Complejidad fisica con {DESCRIPCION}\n"
                       f"{parametros} cadenas={N_CADENAS} "
                       f"semilla_maestra={SEMILLA_MAESTRA}\n"
                       f"intervalo de confianza {int(NIVEL*100)}% con t de "
                       f"Student (t={t_crit:.4f})\n"
                       f"l,C_media,C_desv_estandar,C_error_estandar,"
                       f"C_ic_inferior,C_ic_superior,c_media,c_desv_estandar,"
                       f"c_error_estandar,c_ic_inferior,c_ic_superior,gamma"))

    # --- complejidad: todas las cadenas ---
    columnas = (",".join(f"C_l{l}" for l in longitudes) + ","
                + ",".join(f"c_l{l}" for l in longitudes))
    np.savetxt(os.path.join(CARPETA_CSV,
                            f"Fase6_complejidad_{base}_todas_las_cadenas.csv"),
               np.column_stack([np.asarray(semillas).astype(np.int64),
                                datos["C"], datos["c"]]),
               fmt=["%d"] + ["%.8e"] * (2 * len(longitudes)), delimiter=",",
               comments="# ",
               header=(f"FASE 6. Complejidad de cada cadena. Fila = una "
                       f"semilla\n{parametros}\nsemilla,{columnas}"))


# =====================================================================
# LEER LO QUE YA ESTA GUARDADO  (para no volver a simular)
# =====================================================================
def archivos_de_N(N):
    base = f"{PARAMS_ARCHIVO}_N{N}"
    return [os.path.join(CARPETA_CSV, f"Fase6_{medida}_{base}_{tipo}.csv")
            for medida in ("Md", "complejidad")
            for tipo in ("promedio", "todas_las_cadenas")]


def leer_filas_resumen():
    """Devuelve {N: fila} con las filas de los CSV de resumen que ya existen."""
    filas = {}
    patron = os.path.join(CARPETA_CSV, f"Fase6_resumen_{PARAMS_ARCHIVO}_*valores.csv")
    for ruta in glob.glob(patron):
        for fila in np.loadtxt(ruta, delimiter=",", comments="#", ndmin=2):
            filas[int(fila[0])] = list(fila)
    return filas


def leer_promedios(N):
    """Lee de los CSV los promedios de M(d) y de C_K(l) de un N."""
    base = f"{PARAMS_ARCHIVO}_N{N}"
    Md = np.loadtxt(os.path.join(CARPETA_CSV, f"Fase6_Md_{base}_promedio.csv"),
                    delimiter=",", comments="#")
    C = np.loadtxt(os.path.join(CARPETA_CSV, f"Fase6_complejidad_{base}_promedio.csv"),
                   delimiter=",", comments="#")
    return {"Md": Md[:, 1], "C": C[:, 1], "C_inf": C[:, 4], "C_sup": C[:, 5]}


# =====================================================================
# GRAFICAS DE UN GRUPO (todos sus N en la misma grafica)
# =====================================================================
COLORES = ["#2a78d6", "#eb6834", "#1baf7a", "#eda100"]   # uno por cada N
GRIS = "#52514e"
TINTA = "#0b0b0b"


def terminar_figura(fig, archivo):
    if DESCARGAR_GRAFICAS:
        ruta = os.path.join(CARPETA_GRAFICAS, archivo)
        fig.savefig(ruta, dpi=200)
        print(f"  Guardado: {ruta}", flush=True)
    plt.show()
    plt.close(fig)


def graficar_grupo(nombre, lista_N, resultados, distancias, longitudes,
                   referencia):
    plt.rcParams.update({
        "font.size": 11,
        "axes.edgecolor": GRIS,
        "axes.labelcolor": TINTA,
        "xtick.color": GRIS,
        "ytick.color": GRIS,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })
    subtitulo = (f"m={M}, s={S}, K={K}, T={T}, sin burn-in, "
                 f"{N_CADENAS} semillas por cada N")
    texto_N = ", ".join(str(n) for n in lista_N)

    # --- Grafica 1: promedios de M(d) para cada N ---
    fig, ax = plt.subplots(figsize=(10, 5.5))
    for N, color in zip(lista_N, COLORES):
        ax.plot(distancias, resultados[N]["Md"], color=color, lw=1.8,
                marker="o", ms=3, label=f"N = {N}")
    ax.plot(distancias, referencia, color=GRIS, lw=1.5, ls="--",
            label="Sesgo esperado de una cadena sin correlación")
    ax.set_xlabel("Distancia d")
    ax.set_ylabel("M(d)  [bits]")
    ax.set_title(f"Información mutua promedio — {DESCRIPCION}, N = {texto_N}\n"
                 + subtitulo, loc="left", color=TINTA)
    ax.set_xlim(0, D_MAX + 1)
    ax.set_ylim(bottom=0)
    ax.grid(axis="y", color="#e5e4df", lw=0.8)
    ax.ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
    ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.13),
              ncol=3, fontsize=9)
    fig.tight_layout()
    terminar_figura(fig, f"Fase6_Md_{PARAMS_ARCHIVO}_{nombre}_promedio.png")

    # --- Grafica 2: promedios de C_K(l) contra l para cada N ---
    fig, ax = plt.subplots(figsize=(10, 5.5))
    for N, color in zip(lista_N, COLORES):
        ax.plot(longitudes, resultados[N]["C"], color=color, lw=1.8,
                marker="o", ms=5, label=f"N = {N}")
    ax.set_xlabel("Longitud de la ventana l")
    ax.set_ylabel("$C_K(l)$  [bits]")
    ax.set_title(f"Complejidad física promedio — {DESCRIPCION}, N = {texto_N}\n"
                 + subtitulo, loc="left", color=TINTA)
    ax.set_xticks(longitudes)
    ax.set_ylim(bottom=0)
    ax.grid(axis="y", color="#e5e4df", lw=0.8)
    ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.13),
              ncol=4, fontsize=9)
    fig.tight_layout()
    terminar_figura(fig, f"Fase6_CK_{PARAMS_ARCHIVO}_{nombre}_promedio.png")


# =====================================================================
# FIGURA RESUMEN: TODOS LOS N EN UNA SOLA FIGURA
# =====================================================================
def graficar_resumen(lista_N, todos, longitudes, referencia):
    """Arriba: M(d) en colores (cada columna es un N). Abajo: C_K(l) contra N.
    Los dos paneles comparten el eje de N."""
    ns = np.array(lista_N, dtype=float)
    matriz = np.array([todos[N]["Md"] for N in lista_N]).T   # fila = d, columna = N

    # cada N ocupa una columna; los bordes van a media distancia (escala log)
    razones = np.sqrt(ns[1:] / ns[:-1])
    bordes_N = np.concatenate([[ns[0] / razones[0]], ns[:-1] * razones,
                               [ns[-1] * razones[-1]]])
    bordes_d = np.arange(0.5, D_MAX + 1)

    fig = plt.figure(figsize=(10, 8.6))
    rejilla = fig.add_gridspec(2, 2, width_ratios=[1, 0.025],
                               height_ratios=[1.25, 1], hspace=0.07, wspace=0.03)
    ax_M = fig.add_subplot(rejilla[0, 0])
    ax_C = fig.add_subplot(rejilla[1, 0], sharex=ax_M)
    ax_barra = fig.add_subplot(rejilla[0, 1])

    # --- panel superior: M(d) ---
    mapa = LinearSegmentedColormap.from_list("azules",
                                             plt.cm.Blues(np.linspace(0.04, 1, 256)))
    norma = LogNorm(vmin=matriz[matriz > 0].min(), vmax=matriz.max())
    malla = ax_M.pcolormesh(bordes_N, bordes_d, matriz, cmap=mapa, norm=norma,
                            shading="flat", rasterized=True)
    ax_M.set_ylabel("Distancia d")
    ax_M.set_title("Información mutua y complejidad física promedio al variar N — "
                   f"{DESCRIPCION}\n"
                   f"m={M}, s={S}, K={K}, T={T}, sin burn-in, "
                   f"{N_CADENAS} semillas por cada N",
                   loc="left", color=TINTA)
    plt.setp(ax_M.get_xticklabels(), visible=False)
    barra = fig.colorbar(malla, cax=ax_barra)
    barra.set_label("M(d) promedio  [bits]   (línea naranja: sesgo esperado)")
    sesgo = float(np.mean(referencia))
    barra.ax.axhline(sesgo, color="#eb6834", lw=2)

    # --- panel inferior: C_K(l) ---
    tonos = plt.cm.Blues(np.linspace(0.35, 1, len(longitudes)))
    for i, l in enumerate(longitudes):
        ax_C.fill_between(ns, [todos[N]["C_inf"][i] for N in lista_N],
                          [todos[N]["C_sup"][i] for N in lista_N],
                          color=tonos[i], alpha=0.18, lw=0)
        ax_C.plot(ns, [todos[N]["C"][i] for N in lista_N], color=tonos[i], lw=2,
                  marker="o", ms=3.5, label=f"l = {l}")
    ax_C.set_xscale("log")
    ax_C.set_xlim(bordes_N[0], bordes_N[-1])
    ax_C.set_xticks([5, 10, 20, 50, 100, 200, 500, 1000, 2000, 5000])
    ax_C.xaxis.set_major_formatter(ScalarFormatter())
    ax_C.xaxis.set_minor_locator(NullLocator())
    ax_C.set_ylim(bottom=0)
    ax_C.set_xlabel("Número de agentes N (escala logarítmica)")
    ax_C.set_ylabel("$C_K(l)$  [bits]")
    ax_C.grid(axis="y", color="#e5e4df", lw=0.8)
    ax_C.legend(frameon=False, loc="upper right", ncol=3, fontsize=9,
                title="Longitud de la ventana")

    fig.subplots_adjust(left=0.08, right=0.91, top=0.92, bottom=0.07)
    terminar_figura(fig, f"Fase6_resumen_Md_CK_{PARAMS_ARCHIVO}_N{lista_N[0]}-a-"
                         f"{lista_N[-1]}_{len(lista_N)}valores.png")


# =====================================================================
# PROGRAMA PRINCIPAL
# =====================================================================
if __name__ == "__main__":
    inicio = time.time()

    # --- las 100 semillas (las mismas para todos los N) ---
    semillas = np.random.SeedSequence(SEMILLA_MAESTRA).generate_state(N_CADENAS)
    distancias = np.arange(1, D_MAX + 1)
    longitudes = np.arange(1, L_MAX + 1)

    t_crit = (t_student.ppf(0.5 + NIVEL / 2, N_CADENAS - 1)
              if t_student is not None else T_95)

    # --- linea de referencia: igual para todos los N (solo depende de T) ---
    referencia = sesgo_esperado(K, T - distancias)

    if GUARDAR_CSV:
        os.makedirs(CARPETA_CSV, exist_ok=True)
    if DESCARGAR_GRAFICAS:
        os.makedirs(CARPETA_GRAFICAS, exist_ok=True)
    if not HAY_JOBLIB:
        print("AVISO: joblib no esta instalado -> se usa un solo nucleo "
              "(pip install joblib)")

    resumen = []
    todos = {}   # promedios de todos los N, para la figura resumen
    filas_previas = leer_filas_resumen() if REUTILIZAR_CSV else {}

    for nombre_grupo, lista_N in GRUPOS:
        resultados = {}

        for N in lista_N:
            gamma = K ** M / N

            if (N in filas_previas
                    and all(os.path.exists(r) for r in archivos_de_N(N))):
                promedios = leer_promedios(N)
                resultados[N] = promedios
                todos[N] = promedios
                resumen.append(filas_previas[N])
                print(f"N={N:5d}  gamma={gamma:.5f}  leido de los CSV "
                      f"(no se simulo)", flush=True)
                continue

            t0 = time.time()
            datos = correr_N(N, M, S, K, T, semillas, D_MAX, L_MAX, N_NUCLEOS)

            media_Md = datos["Md"].mean(axis=0)
            media_C = datos["C"].mean(axis=0)
            media_c = datos["c"].mean(axis=0)
            frec = datos["frec"].mean(axis=0)
            tasa_media = datos["tasa_exito"].mean()
            azar = exito_al_azar(N, K)
            d_max = int(distancias[np.argmax(media_Md)])
            resultados[N] = {"Md": media_Md, "C": media_C}
            rC = resumir(datos["C"], t_crit)
            todos[N] = {"Md": media_Md, "C": media_C,
                        "C_inf": rC["ic_inf"], "C_sup": rC["ic_sup"]}

            print(f"N={N:5d}  gamma={gamma:.5f}  "
                  f"frec(0,1,2)=({frec[0]:.3f}, {frec[1]:.3f}, {frec[2]:.3f})  "
                  f"exito={tasa_media:.4f} (azar {azar:.4f})  "
                  f"M(1)={media_Md[0]:.3e}  M_max={media_Md.max():.3e} (d={d_max})  "
                  f"C(1)={media_C[0]:.3e}  C(6)={media_C[5]:.3e}  "
                  f"[{time.time() - t0:.1f} s]", flush=True)

            if GUARDAR_CSV:
                guardar_csv_N(N, gamma, datos, semillas, t_crit, distancias,
                              longitudes, referencia)

            resumen.append([N, gamma, tasa_media, azar,
                            media_Md[0], media_Md[1], media_Md[2], media_Md[3],
                            media_Md.max(), d_max, *media_C, *media_c])

        # --- las graficas del grupo, con todos sus N juntos ---
        print(f"\nGraficas del grupo {nombre_grupo}:")
        graficar_grupo(nombre_grupo, lista_N, resultados, distancias,
                       longitudes, referencia)

    # --- una sola figura con todos los N ---
    print("\nFigura resumen con todos los N:")
    graficar_resumen(LISTA_N, todos, longitudes, referencia)

    # --- resumen de todos los N en un solo archivo ---
    if GUARDAR_CSV:
        nombre_resumen = os.path.join(
            CARPETA_CSV,
            f"Fase6_resumen_{PARAMS_ARCHIVO}_N{LISTA_N[0]}-a-{LISTA_N[-1]}"
            f"_{len(LISTA_N)}valores.csv")
        columnas_C = ",".join(f"C_l{l}" for l in longitudes)
        columnas_c = ",".join(f"c_l{l}" for l in longitudes)
        np.savetxt(nombre_resumen, np.array(resumen),
                   fmt=["%d", "%.6f", "%.6f", "%.6f",
                        "%.8e", "%.8e", "%.8e", "%.8e", "%.8e", "%d"]
                       + ["%.8e"] * (2 * L_MAX),
                   delimiter=",", comments="# ",
                   header=(f"FASE 6. Resumen, {DESCRIPCION}, m={M} s={S} K={K} "
                           f"T={T} cadenas={N_CADENAS}\n"
                           f"N,gamma,tasa_exito,exito_al_azar,"
                           f"M1,M2,M3,M4,M_max,d_del_maximo,"
                           f"{columnas_C},{columnas_c}"))
        print(f"\nGuardado: {nombre_resumen}")

    print(f"\nListo. Tiempo total: {(time.time() - inicio) / 60:.1f} min")
```
