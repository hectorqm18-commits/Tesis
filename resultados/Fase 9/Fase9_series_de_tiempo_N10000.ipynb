```python
"""
FASE 9 (tercera parte) - Precio y series de tiempo, apartado 6.7 del protocolo
==============================================================================
Tesis: Complejidad e informacion en un minority game con opcion de abstencion

Alfabeto del juego: 0 = comprar, 1 = vender, 2 = abstenerse

Que hace este codigo, para cada caso (m, con o sin puntaje):
  1. Simula 100 cadenas y guarda en cada turno N0, N1, N2, el ganador y si
     hubo moneda (mismo nucleo que Fase9_precios, sin cambios).
  2. Calcula tres precios y sus rendimientos r_t (cambio de ln p):
       Mansilla:           p_t = arctan(N0/N1);  r_t = ln p_t - ln p_{t-1}
       Exceso de demanda:  r_t = (N0 - N1) / LAMBDA_EXCESO
       Cociente:           r_t = ln[(N0+1)/(N1+1)] / LAMBDA_COCIENTE
  3. Hechos estilizados de los rendimientos:
       - curtosis (exceso sobre la normal) y exponente de la cola de |r - media|
         (estimador de Hill con la fraccion FRACCION_COLA de valores mas grandes;
         NO es la funcion de Hill de otras fases);
       - autocorrelacion de la volatilidad |r_t| y su exponente de decaimiento.
  4. Fraccion de agentes que se abstienen, f2(t) = N2/N.
  5. Codificacion simbolica de cada precio:
       binaria : 1 si el precio sube, 0 si baja; los cambios nulos se omiten;
       ternaria: 2 si |r_t| < u * (desviacion estandar de r), 1 si sube, 0 si
                 baja, para cada umbral u de UMBRALES.
  6. Medidas informacionales de cada serie simbolica: C(l) con el ajuste
     C(l) = beta*l^alpha, M(d) con su ajuste log-log, y el espectro de potencias
     de M(d), Ec. (4) del protocolo, con el ajuste S ~ 1/f^a.
  Las medidas se calculan por cadena y se promedian sobre las 100 cadenas, sin
  burn-in y con burn-in. Los ajustes se hacen sobre las curvas promedio.

No incluye: indices reales (NASDAQ, IPC) ni burbujas (apartado 6.8).
gamma = K^m / N va en los CSV, no en los titulos.

Sin cambios respecto a otros notebooks: regla_ganador, simular_sin_recompensa,
simular_con_puntaje, informacion_mutua, correr_cadena, correr_caso,
precio_mansilla, rendimiento_exceso, rendimiento_cociente, acumular,
terminar_figura, series_de_precio y graficar_caso (Fase9_precios);
complejidad_fisica (Fase 8); ajuste_loglog (Fase 7) y ajuste_potencia
(Fase9_complejidad_precios).
"""

import os
import time

import numpy as np
import matplotlib.pyplot as plt

try:
    from joblib import Parallel, delayed
    HAY_JOBLIB = True
except ImportError:
    HAY_JOBLIB = False


# =====================================================================
# OPCIONES
# =====================================================================
# Las graficas SIEMPRE se muestran en el notebook.
DESCARGAR_GRAFICAS = True     # True = ademas se guardan como PNG
GUARDAR_CSV        = True     # True = se guardan los CSV
OPCION_VACIA_GANA  = False    # False = regla nueva (una opcion vacia no cuenta)

CARPETA_CSV      = "csv"
CARPETA_GRAFICAS = "graficas"

#   -1 -> usa TODOS los nucleos     1 -> un solo nucleo
N_NUCLEOS = -1


# =====================================================================
# PARAMETROS
# =====================================================================
SEMILLA_MAESTRA = 1907443528
N_CADENAS = 100

N       = 10000  # agentes
S       = 2      # estrategias por agente
K       = 3      # tamano del alfabeto del juego
T       = 6000   # turnos registrados, desde el primero
BURN_IN = 1000   # las medidas se calculan sin burn-in y con burn-in

CASOS = [{"caso": "A", "m": 2, "puntaje": False}, {"caso": "B", "m": 2, "puntaje": True}, {"caso": "C", "m": 3, "puntaje": False}, {"caso": "D", "m": 3, "puntaje": True}, {"caso": "E", "m": 4, "puntaje": True}, {"caso": "F", "m": 5, "puntaje": True}, {"caso": "G", "m": 6, "puntaje": True}]

LAMBDA_EXCESO   = N
LAMBDA_COCIENTE = np.log(N + 1)

PRECIOS  = ["mansilla", "exceso", "cociente"]
UMBRALES = [0.1, 0.25, 0.5]     # codificacion ternaria: fracciones de la desv. estandar
FRACCION_COLA = 0.05            # estimador de Hill: 5 % de los |r - media| mas grandes
LAG_MAX  = 100                  # autocorrelacion de |r| para retardos 1..LAG_MAX
D_MAX    = 100                  # M(d) para d = 1..D_MAX
L_MAX_BIN = 8                   # C(l) de la serie binaria, l = 1..8
L_MAX_TER = 6                   # C(l) de la serie ternaria, l = 1..6
L_MIN_AJUSTE = 2                # el ajuste de C(l) usa l >= 2
ZOOM = 200                      # acercamiento del precio de Mansilla


# =====================================================================
# NUCLEO DEL JUEGO  (igual que en Fase9_precios)
# =====================================================================
def regla_ganador(conteos, rng, opcion_vacia_gana):
    """Devuelve el simbolo ganador y si se decidio con moneda.

    Regla vieja: gana el minimo de las K opciones (una vacia cuenta).
    Regla nueva: gana el minimo entre las opciones con al menos un agente.
    El desempate usa rng.choice igual que el codigo de la Fase 2, asi que
    se consume la misma cantidad de numeros aleatorios en las dos reglas.
    """
    if opcion_vacia_gana:
        minimo = conteos.min()
    else:
        minimo = conteos[conteos > 0].min()
    empatados = np.flatnonzero(conteos == minimo)
    return int(rng.choice(empatados)), len(empatados) > 1


def simular_sin_recompensa(N, m, s, K, T, semilla, opcion_vacia_gana):
    """Corre una partida con agentes que tienen s estrategias pero no
    las califican.

    Todo es igual al codigo base, excepto que se quita la recompensa:
    los puntajes valen 0 siempre. Entonces en el paso 2 las s estrategias
    de cada agente estan empatadas y el ruido decide cual se usa. Es decir,
    en cada ciclo cada agente toma al azar una de sus estrategias y hace
    lo que esa estrategia dice para la historia actual.

    Diferencia clave con los agentes ingenuos: aqui la decision SI depende
    de la memoria, porque las estrategias son tablas que responden a la
    historia de los ultimos m ganadores.
    """
    rng = np.random.default_rng(semilla)
    N_HIST = K ** m

    # --- agentes (igual que el codigo base) ---
    # estrategias[i, j, h] = accion que la estrategia j del agente i
    #                        recomienda ante la historia con indice h
    estrategias = rng.integers(0, K, size=(N, s, N_HIST))

    # puntajes[i, j]: se quedan en 0 porque no hay recompensa
    puntajes = np.zeros((N, s))

    # --- memoria: los m datos que el mercado tiene "al nacer" ---
    historia = rng.integers(0, K, size=m)

    pesos = K ** np.arange(m - 1, -1, -1)   # convierte historia -> indice
    indices = np.arange(N)

    serie_ganadores = np.zeros(T, dtype=int)
    conteos_por_turno = np.zeros((T, K), dtype=np.int32)   # N0, N1, N2
    con_moneda = np.zeros(T, dtype=bool)

    for t in range(T):

        # 1. que situacion estan viendo los agentes
        idx = int(historia @ pesos)

        # 2. cada agente toma su estrategia con mas puntos.
        #    Como todos los puntajes valen 0, el ruido decide al azar.
        mejor = np.argmax(puntajes + rng.random((N, s)) * 0.5, axis=1)

        # 3. consulta esa estrategia en la historia actual y actua
        acciones = estrategias[indices, mejor, idx]

        # 4. se cuenta cuantos eligieron cada opcion
        conteos = np.bincount(acciones, minlength=K)

        # 5. gana la minoria (segun la regla elegida)
        ganador, hubo_moneda = regla_ganador(conteos, rng, opcion_vacia_gana)

        # 6. (SIN RECOMPENSA: en el codigo base aqui se sumaban puntos)

        # 7. registrar desde el primer ciclo
        serie_ganadores[t] = ganador
        conteos_por_turno[t] = conteos
        con_moneda[t] = hubo_moneda

        # 8. la memoria se desplaza: entra el ganador, sale el mas viejo
        historia = np.roll(historia, -1)
        historia[-1] = ganador

    return serie_ganadores, conteos_por_turno, con_moneda


def simular_con_puntaje(N, m, s, K, T, semilla, opcion_vacia_gana):
    """Corre una partida con las reglas del codigo base, sin burn-in.

    Cada agente tiene s estrategias. En cada ciclo usa la que tiene mas
    puntos (los empates se rompen al azar), y al final del ciclo todas
    las estrategias que habrian acertado al ganador reciben un punto.

    Devuelve la serie de ganadores, cuantas veces acerto cada agente, los
    conteos N0, N1, N2 de cada turno y si el turno se decidio con moneda.
    """
    rng = np.random.default_rng(semilla)
    N_HIST = K ** m

    # --- agentes (igual que el codigo base) ---
    # estrategias[i, j, h] = accion que la estrategia j del agente i
    #                        recomienda ante la historia con indice h
    estrategias = rng.integers(0, K, size=(N, s, N_HIST))

    # puntajes[i, j] = puntos virtuales (casos de exito) de esa estrategia
    puntajes = np.zeros((N, s))

    # --- memoria: los m datos que el mercado tiene "al nacer" ---
    historia = rng.integers(0, K, size=m)

    pesos = K ** np.arange(m - 1, -1, -1)   # convierte historia -> indice
    indices = np.arange(N)

    serie_ganadores = np.zeros(T, dtype=int)
    conteos_por_turno = np.zeros((T, K), dtype=np.int32)   # N0, N1, N2
    con_moneda = np.zeros(T, dtype=bool)
    exitos_agente = np.zeros(N, dtype=int)

    for t in range(T):

        # 1. que situacion estan viendo los agentes
        idx = int(historia @ pesos)

        # 2. cada agente toma su estrategia con mas puntos.
        #    El ruido < 1 rompe empates al azar sin alterar el orden
        #    entre puntajes enteros distintos.
        mejor = np.argmax(puntajes + rng.random((N, s)) * 0.5, axis=1)

        # 3. consulta esa estrategia en la historia actual y actua
        acciones = estrategias[indices, mejor, idx]

        # 4. se cuenta cuantos eligieron cada opcion
        conteos = np.bincount(acciones, minlength=K)

        # 5. gana la minoria (segun la regla elegida)
        ganador, hubo_moneda = regla_ganador(conteos, rng, opcion_vacia_gana)

        # 6. se premian TODAS las estrategias que hubieran acertado,
        #    se hayan jugado o no (regla del codigo base / PRE 62).
        puntajes += (estrategias[:, :, idx] == ganador)

        # 7. registrar desde el primer ciclo
        serie_ganadores[t] = ganador
        conteos_por_turno[t] = conteos
        con_moneda[t] = hubo_moneda
        exitos_agente += (acciones == ganador)

        # 8. la memoria se desplaza: entra el ganador, sale el mas viejo
        historia = np.roll(historia, -1)
        historia[-1] = ganador

    return serie_ganadores, exitos_agente, conteos_por_turno, con_moneda


def correr_cadena(N, m, s, K, T, semilla, puntaje, opcion_vacia_gana):
    if puntaje:
        serie, _, conteos, moneda = simular_con_puntaje(N, m, s, K, T, semilla,
                                                        opcion_vacia_gana)
    else:
        serie, conteos, moneda = simular_sin_recompensa(N, m, s, K, T, semilla,
                                                        opcion_vacia_gana)
    return serie, conteos, moneda


def correr_caso(N, m, s, K, T, semillas, puntaje, opcion_vacia_gana, n_nucleos):
    """Corre todas las cadenas. El resultado no depende del numero de nucleos:
    cada cadena usa solo su semilla y joblib las devuelve en orden."""
    if HAY_JOBLIB and n_nucleos != 1:
        bloque = Parallel(n_jobs=n_nucleos)(
            delayed(correr_cadena)(N, m, s, K, T, int(sem), puntaje, opcion_vacia_gana)
            for sem in semillas)
    else:
        bloque = [correr_cadena(N, m, s, K, T, int(sem), puntaje, opcion_vacia_gana)
                  for sem in semillas]
    return {"ganador": np.array([b[0] for b in bloque]),
            "conteos": np.array([b[1] for b in bloque]),
            "moneda": np.array([b[2] for b in bloque])}


# =====================================================================
# PRECIOS  (igual que en Fase9_precios)
# =====================================================================
def precio_mansilla(N0, N1):
    """p_t = arctan(N0/N1), con np.arctan2(N0, N1).

    Nadie vende  (N1 = 0, N0 > 0): pi/2.   Nadie compra (N0 = 0, N1 > 0): 0.
    Nadie compra ni vende (N0 = N1 = 0): se repite el precio anterior
    (si es el primer turno, pi/4).
    """
    p = np.arctan2(N0, N1).astype(float)
    for t in np.flatnonzero((N0 == 0) & (N1 == 0)):
        p[t] = p[t - 1] if t > 0 else np.pi / 4
    return p


def rendimiento_exceso(N0, N1):
    return (N0 - N1) / LAMBDA_EXCESO


def rendimiento_cociente(N0, N1):
    return np.log((N0 + 1) / (N1 + 1)) / LAMBDA_COCIENTE


def acumular(rendimientos, inicio):
    """ln p desde el turno 'inicio', empezando en ln p = 0. Tiene un punto
    mas que turnos: el ultimo es el precio despues del ultimo turno."""
    return np.concatenate([[0.0], np.cumsum(rendimientos[inicio:])])


def rendimientos_y_signo(precio, N0, N1):
    """Rendimiento r_t de un precio y el signo exacto de su cambio (+1, 0, -1).

    Mansilla: r tiene T-1 valores (cambio entre turnos consecutivos). El signo
    se obtiene comparando N0/N1 con enteros, para no depender del redondeo.
    Exceso y cociente: r tiene T valores y el signo es el de N0 - N1.
    """
    if precio == "mansilla":
        p = precio_mansilla(N0.astype(float), N1.astype(float))
        if (p <= 0).any():
            raise RuntimeError("El precio de Mansilla vale 0 en algun turno (N0 = 0): "
                               "su logaritmo no existe. Se detiene para avisar.")
        a, b = N0.copy(), N1.copy()
        for t in np.flatnonzero((a == 0) & (b == 0)):
            a[t], b[t] = (a[t - 1], b[t - 1]) if t > 0 else (1, 1)
        return np.diff(np.log(p)), np.sign(a[1:] * b[:-1] - a[:-1] * b[1:])
    if precio == "exceso":
        return rendimiento_exceso(N0.astype(float), N1.astype(float)), np.sign(N0 - N1)
    return rendimiento_cociente(N0.astype(float), N1.astype(float)), np.sign(N0 - N1)


# =====================================================================
# HECHOS ESTILIZADOS
# =====================================================================
def curtosis(r):
    """Curtosis en exceso: 0 para una distribucion normal, > 0 con colas anchas."""
    x = r - r.mean()
    varianza = np.mean(x ** 2)
    return np.mean(x ** 4) / varianza ** 2 - 3 if varianza > 0 else np.nan


def exponente_cola(r):
    """Estimador de Hill del exponente de la cola de |r - media|.

    Toma los k valores mas grandes (k = FRACCION_COLA del total) y devuelve
    k / sum(ln(x_i / x_(k+1))). Un valor chico indica colas anchas.
    """
    x = np.sort(np.abs(r - r.mean()))[::-1]
    k = max(int(FRACCION_COLA * len(x)), 10)
    if x[k] <= 0:
        return np.nan
    suma = np.sum(np.log(x[:k] / x[k]))
    return k / suma if suma > 0 else np.nan


def autocorrelacion(x, lag_max):
    """Autocorrelacion de x para los retardos 1, ..., lag_max."""
    x = x - x.mean()
    varianza = x @ x
    if varianza == 0:
        return np.full(lag_max, np.nan)
    return np.array([x[:-k] @ x[k:] for k in range(1, lag_max + 1)]) / varianza


# =====================================================================
# MEDIDAS INFORMACIONALES
# =====================================================================
def informacion_mutua(serie, d, K):
    """M(d) = sum_{a,b} P_ab(d) * log2[ P_ab(d) / (P_a * P_b) ]

    P_ab(d): probabilidad de encontrar el simbolo a y, d lugares despues,
             el simbolo b. Se estima contando todos los pares
             (serie[i], serie[i+d]) de la cadena.
    P_a:     densidad del simbolo a en la cadena completa (como dice
             Mansilla: "the density of the symbol alpha").

    La suma corre sobre los K*K = 9 pares (a, b). Los pares que nunca
    aparecen (P_ab = 0) no aportan, porque 0 * log(0) = 0.
    """
    x = serie[:-d]          # simbolo alpha
    y = serie[d:]           # simbolo beta, d lugares despues
    n_pares = len(x)

    # probabilidades conjuntas P_ab(d): matriz K x K
    conjunta = np.bincount(x * K + y, minlength=K * K).reshape(K, K) / n_pares

    # densidades P_a de cada simbolo en toda la cadena
    densidad = np.bincount(serie, minlength=K) / len(serie)
    producto = np.outer(densidad, densidad)      # P_a * P_b

    M_d = 0.0
    for a in range(K):
        for b in range(K):
            if conjunta[a, b] > 0:
                M_d += conjunta[a, b] * np.log2(conjunta[a, b] / producto[a, b])
    return M_d


def complejidad_fisica(serie, l, K):
    """Devuelve C_K(l) y c_K(l) de una cadena de ganadores.

    Una ventana de longitud l recorre la cadena de uno en uno. Cada ventana
    se convierte en un numero (en base K) para contar cuantas veces aparece
    cada subcadena eta. Con esas frecuencias relativas p(eta):

        H(l)   = -sum p(eta) log2 p(eta)
        C_K(l) = l*log2(K) - H(l)
        c_K(l) = C_K(l) / (l*log2(K))
    """
    ventanas = np.lib.stride_tricks.sliding_window_view(serie, l)
    pesos = K ** np.arange(l - 1, -1, -1)
    codigos = ventanas @ pesos
    conteos = np.bincount(codigos, minlength=K ** l)

    p = conteos[conteos > 0] / len(codigos)
    H = -np.sum(p * np.log2(p))

    maximo = l * np.log2(K)
    C = maximo - H
    return C, C / maximo


def ajuste_loglog(d, Md):
    """Quita los ceros, pasa a log10 y ajusta una recta.

    Devuelve un diccionario con la pendiente, la ordenada al origen,
    la R^2, los puntos usados y cuantos ceros se quitaron.
    """
    validos = Md > 0                      # UNICO filtro: fuera M(d) = 0
    x = np.log10(d[validos])
    y = np.log10(Md[validos])

    pendiente, ordenada = np.polyfit(x, y, 1)   # recta y = ordenada + pendiente x
    y_ajuste = ordenada + pendiente * x

    ss_res = np.sum((y - y_ajuste) ** 2)        # suma de cuadrados residual
    ss_tot = np.sum((y - y.mean()) ** 2)        # suma de cuadrados total
    r2 = 1 - ss_res / ss_tot

    return {
        "pendiente": pendiente,
        "ordenada": ordenada,
        "r2": r2,
        "x": x, "y": y, "y_ajuste": y_ajuste,
        "n_puntos": int(validos.sum()),
        "n_ceros": int((~validos).sum()),
    }


def ajuste_potencia(longitudes, C):
    """Ajuste de C(l) = beta * l^alpha: recta en log10 C contra log10 l."""
    usar = (longitudes >= L_MIN_AJUSTE) & (C > 0)
    x, y = np.log10(longitudes[usar]), np.log10(C[usar])
    alpha, ordenada = np.polyfit(x, y, 1)
    y_ajuste = alpha * x + ordenada
    r2 = 1 - np.sum((y - y_ajuste) ** 2) / np.sum((y - y.mean()) ** 2)
    return {"alpha": alpha, "beta": 10 ** ordenada, "r2": r2, "x": x, "y": y,
            "y_ajuste": y_ajuste, "puntos": int(usar.sum())}


def recta_segura(x, y):
    """ajuste_loglog; si hay menos de 3 puntos positivos no ajusta y devuelve nan."""
    if (y > 0).sum() < 3:
        return {"pendiente": np.nan, "r2": np.nan, "n_puntos": int((y > 0).sum())}
    return ajuste_loglog(x, y)


def potencia_segura(longitudes, C):
    """ajuste_potencia; si hay menos de 3 puntos utiles no ajusta y devuelve nan."""
    if ((longitudes >= L_MIN_AJUSTE) & (C > 0)).sum() < 3:
        return {"alpha": np.nan, "beta": np.nan, "r2": np.nan}
    return ajuste_potencia(longitudes, C)


def texto(v):
    """Formato para el CSV: enteros tal cual, decimales con 8 cifras."""
    return str(int(v)) if isinstance(v, (int, np.integer)) else f"{v:.8e}"


def espectro_potencias(Md):
    """Ec. (4) del protocolo con u = 1: S(k) = |sum_d M(d) exp(-i 2 pi (k/L) d)|^2,
    para k = 1, ..., L/2. Devuelve la frecuencia k/L y S(k)."""
    L = len(Md)
    d = np.arange(1, L + 1)
    k = np.arange(1, L // 2 + 1)
    return k / L, np.abs(np.exp(-2j * np.pi * np.outer(k, d) / L) @ Md) ** 2


def codificaciones(r, signo):
    """Devuelve las series simbolicas de un precio: la binaria y una ternaria
    por cada umbral. Cada elemento es (nombre, umbral, alfabeto, serie)."""
    series = [("binaria", 0.0, 2, (signo[signo != 0] > 0).astype(np.int64))]
    desv = r.std()
    for u in UMBRALES:
        ternaria = np.where(np.abs(r) < u * desv, 2, (r > 0).astype(np.int64))
        series.append((f"ternaria_{u}", u, 3, ternaria.astype(np.int64)))
    return series


NOMBRES_COD = ["binaria"] + [f"ternaria_{u}" for u in UMBRALES]


def analizar_cadena(conteos):
    """Todas las medidas de una cadena, para cada precio, sin y con burn-in."""
    N0 = conteos[:, 0].astype(np.int64)
    N1 = conteos[:, 1].astype(np.int64)
    salida = {}
    for precio in PRECIOS:
        r_todo, signo_todo = rendimientos_y_signo(precio, N0, N1)
        for inicio in (0, BURN_IN):
            r, signo = r_todo[inicio:], signo_todo[inicio:]
            medidas = {"media": r.mean(), "desv": r.std(), "curtosis": curtosis(r),
                       "cola": exponente_cola(r),
                       "acf": autocorrelacion(np.abs(r), LAG_MAX)}
            for nombre, _, alfabeto, serie in codificaciones(r, signo):
                l_max = L_MAX_BIN if alfabeto == 2 else L_MAX_TER
                C = np.full(L_MAX_BIN, np.nan)
                C[:l_max] = [complejidad_fisica(serie, l, alfabeto)[0]
                             for l in range(1, l_max + 1)]
                medidas[nombre] = {
                    "largo": len(serie),
                    "frec": np.bincount(serie, minlength=3) / len(serie),
                    "C": C,
                    "Md": np.array([informacion_mutua(serie, d, alfabeto)
                                    for d in range(1, D_MAX + 1)])}
            salida[(precio, inicio)] = medidas
    return salida


def analizar_caso(conteos, n_nucleos):
    if HAY_JOBLIB and n_nucleos != 1:
        return Parallel(n_jobs=n_nucleos)(delayed(analizar_cadena)(c) for c in conteos)
    return [analizar_cadena(c) for c in conteos]


def promediar(analisis, precio, inicio):
    """Promedios sobre las cadenas y ajustes sobre las curvas promedio."""
    cadenas = [a[(precio, inicio)] for a in analisis]
    lags = np.arange(1, LAG_MAX + 1)
    distancias = np.arange(1, D_MAX + 1)
    longitudes = np.arange(1, L_MAX_BIN + 1)
    acf = np.nanmean([c["acf"] for c in cadenas], axis=0)
    cola = np.array([c["cola"] for c in cadenas])
    p = {"media": np.mean([c["media"] for c in cadenas]),
         "desv": np.mean([c["desv"] for c in cadenas]),
         "curtosis": np.nanmean([c["curtosis"] for c in cadenas]),
         "cola": np.nanmean(cola) if np.isfinite(cola).any() else np.nan,
         "cola_validas": int(np.isfinite(cola).sum()),
         "acf": acf, "acf_ajuste": recta_segura(lags, acf), "cod": {}}
    for nombre in NOMBRES_COD:
        C = np.mean([c[nombre]["C"] for c in cadenas], axis=0)
        Md = np.mean([c[nombre]["Md"] for c in cadenas], axis=0)
        validos = np.isfinite(C)
        frec, S = espectro_potencias(Md)
        p["cod"][nombre] = {
            "largo": np.mean([c[nombre]["largo"] for c in cadenas]),
            "frec": np.mean([c[nombre]["frec"] for c in cadenas], axis=0),
            "C": C, "Md": Md, "f": frec, "S": S,
            "aj_C": potencia_segura(longitudes[validos], C[validos]),
            "aj_M": recta_segura(distancias, Md),
            "aj_S": recta_segura(frec, S)}
    return p


# =====================================================================
# GRAFICAS
# =====================================================================
AZUL = "#2a78d6"
GRIS = "#52514e"
TINTA = "#0b0b0b"
NOMBRES_PRECIO = ["Mansilla", "Exceso de demanda", "Cociente"]
COLORES_COD = dict(zip(NOMBRES_COD, ["#2a78d6", "#eb6834", "#1baf7a", "#eda100"]))
ETIQUETAS_COD = dict(zip(NOMBRES_COD, ["Binaria"] + [f"Ternaria, umbral {u}" for u in UMBRALES]))
COLORES_PRECIO = dict(zip(PRECIOS, ["#2a78d6", "#eb6834", "#1baf7a"]))
MARCAS_PRECIO = dict(zip(PRECIOS, ["o", "s", "^"]))


def terminar_figura(fig, archivo):
    if DESCARGAR_GRAFICAS:
        ruta = os.path.join(CARPETA_GRAFICAS, archivo)
        fig.savefig(ruta, dpi=150)
        print(f"  Guardado: {ruta}", flush=True)
    plt.show()
    plt.close(fig)


def series_de_precio(conteos_cadena, inicio):
    """Los tres precios de una cadena desde el turno 'inicio', con su eje de turnos."""
    N0 = conteos_cadena[:, 0].astype(float)
    N1 = conteos_cadena[:, 1].astype(float)
    turnos = np.arange(inicio, len(N0))
    turnos_acum = np.arange(inicio, len(N0) + 1)
    return [(turnos, precio_mansilla(N0, N1)[inicio:]),
            (turnos_acum, acumular(rendimiento_exceso(N0, N1), inicio)),
            (turnos_acum, acumular(rendimiento_cociente(N0, N1), inicio))]


def graficar_caso(caso, datos, semillas, destacada, inicio, base):
    texto_puntaje = "con puntaje" if caso["puntaje"] else "sin puntaje"
    if inicio:
        texto_burn = f"con burn-in de {inicio} turnos (turnos {inicio} a {T - 1})"
        sufijo = f"con_burnin{inicio}"
    else:
        texto_burn = f"sin burn-in (turnos 0 a {T - 1})"
        sufijo = "sin_burnin"
    encabezado = (f"Caso {caso['caso']} — {texto_puntaje}, N={N}, m={caso['m']}, s={S}, "
                  f"{texto_burn}")
    etiquetas_y = ["p  (Mansilla)", "ln p  (exceso de demanda)", "ln p  (cociente)"]
    etiqueta_dest = f"Cadena destacada (semilla {int(semillas[destacada])})"

    todas = [series_de_precio(datos["conteos"][c], inicio) for c in range(N_CADENAS)]

    # --- a) conjunto: las 100 cadenas y la destacada ---
    fig, ejes = plt.subplots(3, 1, figsize=(11, 11), sharex=True)
    for j, ax in enumerate(ejes):
        for c in range(N_CADENAS):
            ax.plot(*todas[c][j], color=GRIS, lw=0.4, alpha=0.25, rasterized=True)
        ax.plot(*todas[destacada][j], color=AZUL, lw=1.3, label=etiqueta_dest)
        if j == 0:
            ax.axhline(np.pi / 4, color=TINTA, lw=1, ls="--", label=r"$\pi/4$")
        ax.set_ylabel(etiquetas_y[j])
        ax.set_title(NOMBRES_PRECIO[j], loc="left", color=TINTA, fontsize=11)
        ax.grid(axis="y", color="#e5e4df", lw=0.8)
    ejes[0].plot([], [], color=GRIS, lw=0.8, alpha=0.6, label=f"Las {N_CADENAS} cadenas")
    ejes[0].legend(frameon=False, loc="upper right", ncol=3, fontsize=9)
    ejes[-1].set_xlabel("Turno")
    fig.suptitle(f"Precios, las {N_CADENAS} cadenas\n{encabezado}", x=0.06, ha="left",
                 color=TINTA, fontsize=12)
    fig.tight_layout()
    terminar_figura(fig, f"Fase9_precios_conjunto_{base}_{sufijo}.png")

    # --- b) detalle de la cadena destacada ---
    serie = todas[destacada]
    fig, ejes = plt.subplots(2, 2, figsize=(13, 8))
    ax = ejes[0, 0]
    ax.plot(*serie[0], color=AZUL, lw=0.5)
    ax.set_title("Mansilla, completo", loc="left", color=TINTA, fontsize=11)
    ax.set_ylabel(etiquetas_y[0])
    ax = ejes[0, 1]
    ax.plot(serie[0][0][:ZOOM], serie[0][1][:ZOOM], color=AZUL, lw=1.2, marker="o", ms=2.5)
    ax.set_title(f"Mansilla, acercamiento de {ZOOM} turnos "
                 f"({inicio} a {inicio + ZOOM - 1})", loc="left", color=TINTA, fontsize=11)
    ax.set_ylabel(etiquetas_y[0])
    for ax in ejes[0]:
        ax.axhline(np.pi / 4, color=TINTA, lw=1, ls="--", label=r"$\pi/4$")
        ax.legend(frameon=False, fontsize=9, loc="upper right")
    for j, ax in ((1, ejes[1, 0]), (2, ejes[1, 1])):
        ax.plot(*serie[j], color=AZUL, lw=1.2)
        ax.set_title(NOMBRES_PRECIO[j], loc="left", color=TINTA, fontsize=11)
        ax.set_ylabel(etiquetas_y[j])
    for ax in ejes.ravel():
        ax.set_xlabel("Turno")
        ax.grid(axis="y", color="#e5e4df", lw=0.8)
    fig.suptitle(f"Precios de la cadena destacada (semilla {int(semillas[destacada])})\n"
                 f"{encabezado}", x=0.06, ha="left", color=TINTA, fontsize=12)
    fig.tight_layout()
    terminar_figura(fig, f"Fase9_precios_detalle_{base}_{sufijo}.png")


def texto_caso(caso):
    return (f"Caso {caso['caso']} — {'con' if caso['puntaje'] else 'sin'} puntaje, "
            f"N={N}, m={caso['m']}, s={S}")


def graficar_series(caso, conteos_cadena, semilla, base):
    """Cadena destacada, con burn-in: r_t, |r_t| de cada precio y f2(t)."""
    N0 = conteos_cadena[:, 0].astype(np.int64)
    N1 = conteos_cadena[:, 1].astype(np.int64)
    fig = plt.figure(figsize=(14, 9.5))
    rejilla = fig.add_gridspec(3, 3)
    for j, precio in enumerate(PRECIOS):
        r, _ = rendimientos_y_signo(precio, N0, N1)
        r = r[BURN_IN:]
        turnos = np.arange(BURN_IN, BURN_IN + len(r))
        ax = fig.add_subplot(rejilla[0, j])
        ax.plot(turnos, r, color=AZUL, lw=0.4)
        ax.set_title(NOMBRES_PRECIO[j], loc="left", color=TINTA, fontsize=11)
        ax.set_ylabel("Rendimiento $r_t$")
        ax = fig.add_subplot(rejilla[1, j])
        ax.plot(turnos, np.abs(r), color=AZUL, lw=0.4)
        ax.set_ylabel("Volatilidad $|r_t|$")
    ax = fig.add_subplot(rejilla[2, :])
    ax.plot(np.arange(BURN_IN, T), conteos_cadena[BURN_IN:, 2] / N, color=AZUL, lw=0.4)
    ax.set_ylabel("Fracción que se abstiene $f_2(t)$")
    for ax in fig.axes:
        ax.set_xlabel("Turno")
        ax.grid(axis="y", color="#e5e4df", lw=0.8)
    fig.suptitle(f"Rendimientos, volatilidad y abstención de la cadena destacada "
                 f"(semilla {int(semilla)})\n{texto_caso(caso)}, con burn-in de {BURN_IN} "
                 f"turnos (turnos {BURN_IN} a {T - 1})", x=0.06, ha="left", color=TINTA,
                 fontsize=12)
    fig.tight_layout()
    terminar_figura(fig, f"Fase9_series_{base}_con_burnin{BURN_IN}.png")


def graficar_hechos(caso, conteos, promedios, base):
    """Con burn-in: distribucion de los rendimientos y autocorrelacion de |r|."""
    fig, ejes = plt.subplots(2, 3, figsize=(14, 8.5))
    z = np.linspace(-6, 6, 241)
    for j, precio in enumerate(PRECIOS):
        p = promedios[(precio, BURN_IN)]
        estandar = []
        for c in conteos:
            r, _ = rendimientos_y_signo(precio, c[:, 0].astype(np.int64),
                                        c[:, 1].astype(np.int64))
            r = r[BURN_IN:]
            if r.std() > 0:
                estandar.append((r - r.mean()) / r.std())
        estandar = np.concatenate(estandar)
        ax = ejes[0, j]
        ax.hist(estandar, bins=121, range=(-8, 8), density=True, color=AZUL, alpha=0.85,
                label=f"Las {N_CADENAS} cadenas juntas")
        ax.plot(z, np.exp(-z ** 2 / 2) / np.sqrt(2 * np.pi), color=TINTA, lw=1.2, ls="--",
                label="Distribución normal")
        ax.set_yscale("log")
        ax.set_ylim(bottom=1e-6)
        ax.set_xlabel("Rendimiento estandarizado  $(r - \\bar r)/\\sigma_r$")
        ax.set_ylabel("Densidad")
        ax.set_title(f"{NOMBRES_PRECIO[j]}\ncurtosis = {p['curtosis']:.3f},  "
                     f"exponente de la cola = {p['cola']:.2f}", loc="left", color=TINTA,
                     fontsize=10.5)
        ax.legend(frameon=False, fontsize=8.5, loc="upper right")
        ax = ejes[1, j]
        ax.plot(np.arange(1, LAG_MAX + 1), p["acf"], color=AZUL, lw=1.3, marker="o", ms=2.5)
        ax.axhline(0, color=GRIS, lw=0.8)
        aj = p["acf_ajuste"]
        nota = (f"exponente de decaimiento = {-aj['pendiente']:.3f},  R$^2$ = {aj['r2']:.3f}\n"
                f"({aj['n_puntos']} retardos con valor positivo)" if np.isfinite(aj["pendiente"])
                else "sin ajuste: menos de 3 retardos con valor positivo")
        ax.set_title(f"Autocorrelación de $|r_t|$\n{nota}", loc="left", color=TINTA,
                     fontsize=10.5)
        ax.set_xlabel("Retardo (turnos)")
        ax.set_ylabel("Autocorrelación")
    for ax in ejes.ravel():
        ax.grid(axis="y", color="#e5e4df", lw=0.8)
    fig.suptitle(f"Hechos estilizados de los rendimientos, promedio de {N_CADENAS} cadenas\n"
                 f"{texto_caso(caso)}, con burn-in de {BURN_IN} turnos", x=0.06, ha="left",
                 color=TINTA, fontsize=12)
    fig.tight_layout()
    terminar_figura(fig, f"Fase9_hechos_estilizados_{base}_con_burnin{BURN_IN}.png")


def graficar_informacion(caso, promedios, base):
    """Con burn-in: C(l), M(d) y espectro de M(d) de las series simbolicas."""
    fig, ejes = plt.subplots(3, 3, figsize=(15, 12.5))
    for j, precio in enumerate(PRECIOS):
        p = promedios[(precio, BURN_IN)]
        for nombre in NOMBRES_COD:
            c = p["cod"][nombre]
            color = COLORES_COD[nombre]
            validos = np.isfinite(c["C"])
            ejes[0, j].plot(np.arange(1, L_MAX_BIN + 1)[validos], c["C"][validos], color=color,
                            marker="o", ms=4, lw=1.5,
                            label=f"{ETIQUETAS_COD[nombre]}:  $\\alpha$ = {c['aj_C']['alpha']:.2f}")
            ejes[1, j].plot(np.arange(1, D_MAX + 1), c["Md"], color=color, lw=1.2,
                            label=(f"{ETIQUETAS_COD[nombre]}:  pendiente = "
                                   f"{c['aj_M']['pendiente']:.2f}, R$^2$ = {c['aj_M']['r2']:.2f}"))
            ejes[2, j].plot(c["f"], c["S"], color=color, lw=1.2,
                            label=(f"{ETIQUETAS_COD[nombre]}:  a = "
                                   f"{-c['aj_S']['pendiente']:.2f}, R$^2$ = {c['aj_S']['r2']:.2f}"))
        ejes[0, j].set_title(f"{NOMBRES_PRECIO[j]}\nComplejidad física C(l)", loc="left",
                             color=TINTA, fontsize=10.5)
        ejes[0, j].set_xlabel("Longitud de la ventana l")
        ejes[0, j].set_ylabel("C(l)  [bits]")
        ejes[1, j].set_title("Información mutua M(d)", loc="left", color=TINTA, fontsize=10.5)
        ejes[1, j].set_xlabel("Distancia d")
        ejes[1, j].set_ylabel("M(d)  [bits]")
        ejes[1, j].ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
        ejes[2, j].set_title("Espectro de potencias de M(d)", loc="left", color=TINTA,
                             fontsize=10.5)
        ejes[2, j].set_xscale("log")
        ejes[2, j].set_yscale("log")
        ejes[2, j].set_xlabel("Frecuencia k/L")
        ejes[2, j].set_ylabel("S(k)")
    for ax in ejes.ravel():
        ax.grid(color="#e5e4df", lw=0.8)
        ax.legend(frameon=False, fontsize=7.5)
    fig.suptitle(f"Medidas informacionales de los precios codificados, promedio de "
                 f"{N_CADENAS} cadenas\n{texto_caso(caso)}, con burn-in de {BURN_IN} turnos",
                 x=0.06, ha="left", color=TINTA, fontsize=12)
    fig.tight_layout()
    terminar_figura(fig, f"Fase9_medidas_informacionales_{base}_con_burnin{BURN_IN}.png")


def graficar_comparacion(todos, que):
    """Los 7 casos juntos, con burn-in. 'que' = 'hechos' o 'informacion'."""
    letras = [c["caso"] for c in CASOS]
    etiquetas = [f"{c['caso']}\nm={c['m']}\n{'con' if c['puntaje'] else 'sin'} p." for c in CASOS]
    if que == "hechos":
        paneles = [("Curtosis de los rendimientos", lambda p: p["curtosis"]),
                   ("Exponente de la cola", lambda p: p["cola"]),
                   ("Autocorrelación de $|r_t|$ con retardo 1", lambda p: p["acf"][0]),
                   ("Exponente de decaimiento de la autocorrelación",
                    lambda p: -p["acf_ajuste"]["pendiente"])]
        titulo = "Hechos estilizados de los rendimientos"
    else:
        bin_ = lambda p: p["cod"]["binaria"]
        paneles = [("Exponente $\\alpha$ de C(l), serie binaria", lambda p: bin_(p)["aj_C"]["alpha"]),
                   (f"C({L_MAX_BIN}) de la serie binaria  [bits]", lambda p: bin_(p)["C"][-1]),
                   ("Pendiente log-log de M(d), serie binaria", lambda p: bin_(p)["aj_M"]["pendiente"]),
                   ("Exponente a del espectro de M(d), serie binaria",
                    lambda p: -bin_(p)["aj_S"]["pendiente"])]
        titulo = "Medidas informacionales de la serie binaria"
    fig, ejes = plt.subplots(2, 2, figsize=(13, 9))
    for ax, (nombre, valor) in zip(ejes.ravel(), paneles):
        for precio, etq in zip(PRECIOS, NOMBRES_PRECIO):
            ax.plot(range(len(letras)), [valor(todos[l][(precio, BURN_IN)]) for l in letras],
                    color=COLORES_PRECIO[precio], marker=MARCAS_PRECIO[precio], ms=7, lw=1.6,
                    label=etq)
        ax.set_xticks(range(len(letras)))
        ax.set_xticklabels(etiquetas, fontsize=9)
        ax.set_title(nombre, loc="left", color=TINTA, fontsize=11)
        ax.grid(axis="y", color="#e5e4df", lw=0.8)
        ax.legend(frameon=False, fontsize=9)
    fig.suptitle(f"{titulo}: comparación de los {len(CASOS)} casos\nN={N}, s={S}, T={T}, "
                 f"{N_CADENAS} cadenas por caso, con burn-in de {BURN_IN} turnos",
                 x=0.06, ha="left", color=TINTA, fontsize=12)
    fig.tight_layout()
    nombre = "hechos_estilizados" if que == "hechos" else "medidas_informacionales"
    terminar_figura(fig, f"Fase9_comparacion_{nombre}_N{N}_s{S}_T{T}_{REGLA}.png")


# =====================================================================
# PROGRAMA PRINCIPAL
# =====================================================================
REGLA = "regla_vieja" if OPCION_VACIA_GANA else "regla_nueva"

if __name__ == "__main__":
    inicio_reloj = time.time()
    semillas = np.random.SeedSequence(SEMILLA_MAESTRA).generate_state(N_CADENAS)

    if GUARDAR_CSV:
        os.makedirs(CARPETA_CSV, exist_ok=True)
    if DESCARGAR_GRAFICAS:
        os.makedirs(CARPETA_GRAFICAS, exist_ok=True)
    plt.rcParams.update({"font.size": 11, "axes.edgecolor": GRIS,
                         "axes.labelcolor": TINTA, "xtick.color": GRIS,
                         "ytick.color": GRIS, "axes.spines.top": False,
                         "axes.spines.right": False})

    print(f"CORRIDA: {REGLA}, N={N}, s={S}, K={K}, T={T}, {N_CADENAS} cadenas, "
          f"burn-in={BURN_IN}", flush=True)

    resumen_cadenas = []      # una fila por caso y cadena
    filas = []                # una fila por caso, precio, burn-in y codificacion
    todos = {}                # promedios, para las graficas de comparacion
    for caso in CASOS:
        m, puntaje = caso["m"], caso["puntaje"]
        texto_puntaje = "con_puntaje" if puntaje else "sin_puntaje"
        base = f"caso{caso['caso']}_N{N}_m{m}_s{S}_{texto_puntaje}_T{T}_{REGLA}"
        gamma = K ** m / N
        t0 = time.time()
        datos = correr_caso(N, m, S, K, T, semillas, puntaje, OPCION_VACIA_GANA, N_NUCLEOS)
        conteos = datos["conteos"]
        N0, N1 = conteos[:, :, 0], conteos[:, :, 1]
        t_sim = time.time() - t0

        # --- turnos especiales y precios acumulados, por cadena ---
        vacia = (conteos == 0).any(axis=2).sum(axis=1)
        sin_ordenes = ((N0 == 0) & (N1 == 0)).sum(axis=1)
        todos_igual = (conteos.max(axis=2) == N).sum(axis=1)
        con_moneda = datos["moneda"].sum(axis=1)
        f2 = conteos[:, :, 2] / N
        recorrido = np.zeros(N_CADENAS)
        filas_caso = []
        for c in range(N_CADENAS):
            n0, n1 = N0[c].astype(float), N1[c].astype(float)
            r_exc, r_coc = rendimiento_exceso(n0, n1), rendimiento_cociente(n0, n1)
            exc, coc = acumular(r_exc, 0), acumular(r_coc, 0)
            exc_burn = acumular(r_exc, BURN_IN)
            recorrido[c] = exc_burn.max() - exc_burn.min()
            filas_caso.append([caso["caso"], m, int(puntaje), N, gamma, c, int(semillas[c]),
                               vacia[c], (N0[c] == 0).sum(), (N1[c] == 0).sum(),
                               sin_ordenes[c], todos_igual[c], con_moneda[c],
                               exc[-1], exc.max(), exc.min(), coc[-1], coc.max(), coc.min(),
                               recorrido[c], f2[c].mean(), f2[c].std()])
        destacada = int(np.argmax(recorrido))
        for c, fila in enumerate(filas_caso):
            fila.append(int(c == destacada))
        resumen_cadenas.extend(filas_caso)

        # --- todas las medidas del apartado 6.7 ---
        analisis = analizar_caso(conteos, N_NUCLEOS)
        promedios = {(precio, inicio): promediar(analisis, precio, inicio)
                     for precio in PRECIOS for inicio in (0, BURN_IN)}
        todos[caso["caso"]] = promedios
        iguales = all(np.array_equal(a[("exceso", i)]["binaria"]["Md"],
                                     a[("cociente", i)]["binaria"]["Md"])
                      for a in analisis for i in (0, BURN_IN))

        print(f"\nCaso {caso['caso']} (m={m}, {texto_puntaje.replace('_', ' ')}), gamma={gamma:.5f}: "
              f"opcion vacia={vacia.sum()}  N0=N1=0: {sin_ordenes.sum()}  todos igual: "
              f"{todos_igual.sum()}  moneda={con_moneda.sum()} de {N_CADENAS * T} turnos.  "
              f"f2 media={f2.mean():.4f}.  Destacada: semilla {int(semillas[destacada])}.  "
              f"Binaria de exceso == binaria de cociente: {iguales}.  "
              f"[simulacion {t_sim:.0f} s, total {time.time() - t0:.0f} s]", flush=True)
        for precio in PRECIOS:
            p = promedios[(precio, BURN_IN)]
            b = p["cod"]["binaria"]
            aj = p["acf_ajuste"]
            print(f"  {precio:9s} (con burn-in): curtosis={p['curtosis']:8.3f}  cola={p['cola']:6.2f}  "
                  f"acf(1)={p['acf'][0]:+.3f}  exp. acf={-aj['pendiente']:6.3f}  |  "
                  f"binaria: alpha={b['aj_C']['alpha']:.3f}  C({L_MAX_BIN})={b['C'][-1]:.3e}  "
                  f"M(1)={b['Md'][0]:.3e}  pend. M={b['aj_M']['pendiente']:+.3f}  "
                  f"a espectro={-b['aj_S']['pendiente']:+.3f}", flush=True)

        for precio in PRECIOS:
            for inicio in (0, BURN_IN):
                p = promedios[(precio, inicio)]
                aj = p["acf_ajuste"]
                for nombre, u in zip(NOMBRES_COD, [0.0] + UMBRALES):
                    c = p["cod"][nombre]
                    filas.append(
                        [caso["caso"], str(m), str(int(puntaje)), str(N), f"{gamma:.6f}", precio,
                         str(inicio)]
                        + [texto(v) for v in (
                            p["media"], p["desv"], p["curtosis"], p["cola"], p["acf"][0],
                            p["acf"][9], p["acf"][99], -aj["pendiente"], aj["r2"],
                            aj["n_puntos"], f2.mean(), f2.std())]
                        + [nombre.split("_")[0], f"{u}"]
                        + [texto(v) for v in (
                            c["largo"], *c["frec"], *c["C"], c["aj_C"]["alpha"], c["aj_C"]["beta"],
                            c["aj_C"]["r2"], c["Md"][0], c["Md"][1], c["Md"][2], c["Md"].max(),
                            int(np.argmax(c["Md"]) + 1), c["aj_M"]["pendiente"], c["aj_M"]["r2"],
                            -c["aj_S"]["pendiente"], c["aj_S"]["r2"])])

        # --- CSV de conteos del caso, formato largo ---
        if GUARDAR_CSV:
            tabla = np.column_stack([
                np.repeat(np.arange(N_CADENAS), T), np.repeat(semillas.astype(np.int64), T),
                np.tile(np.arange(T), N_CADENAS), conteos.reshape(-1, K),
                datos["ganador"].ravel(), datos["moneda"].ravel().astype(np.int64)])
            np.savetxt(os.path.join(CARPETA_CSV, f"Fase9_conteos_{base}.csv"), tabla,
                       fmt="%d", delimiter=",", comments="",
                       header="cadena,semilla,turno,N0,N1,N2,ganador,moneda")

        # --- graficas del caso ---
        print(f"Graficas del caso {caso['caso']}:")
        for inicio in (0, BURN_IN):
            graficar_caso(caso, datos, semillas, destacada, inicio, base)
        graficar_series(caso, conteos[destacada], semillas[destacada], base)
        graficar_hechos(caso, conteos, promedios, base)
        graficar_informacion(caso, promedios, base)
        del datos, conteos, analisis

    print("\nGraficas de comparacion:")
    graficar_comparacion(todos, "hechos")
    graficar_comparacion(todos, "informacion")

    if GUARDAR_CSV:
        ruta = os.path.join(CARPETA_CSV, f"resumen_fase9_N{N}.csv")
        with open(ruta, "w", encoding="utf8") as archivo:
            archivo.write(f"# FASE 9. {REGLA}, N={N} s={S} K={K} T={T} burn_in_graficas={BURN_IN} "
                          f"semilla_maestra={SEMILLA_MAESTRA}\n"
                          "# Los valores de ln p y de f2 son de toda la serie (sin burn-in).\n"
                          "# Cadena destacada: mayor recorrido (maximo - minimo) de ln p con exceso "
                          f"de demanda despues del burn-in (turnos {BURN_IN} a {T - 1}).\n"
                          "caso,m,puntaje,N,gamma,cadena,semilla,turnos_opcion_vacia,turnos_N0_cero,"
                          "turnos_N1_cero,turnos_N0_y_N1_cero,turnos_todos_misma_opcion,"
                          "turnos_con_moneda,lnp_exceso_final,lnp_exceso_max,lnp_exceso_min,"
                          "lnp_cociente_final,lnp_cociente_max,lnp_cociente_min,"
                          "recorrido_exceso_con_burnin,f2_media,f2_desv,destacada\n")
            for f in resumen_cadenas:
                archivo.write(",".join([f[0], str(f[1]), str(f[2]), str(f[3]), f"{f[4]:.6f}"]
                                       + [str(int(v)) for v in f[5:13]]
                                       + [f"{v:.8e}" for v in f[13:22]] + [str(f[22])]) + "\n")
        print(f"\nGuardado: {ruta}")

        ruta = os.path.join(CARPETA_CSV, f"Fase9_parametros_6_7_N{N}_s{S}_T{T}_{REGLA}.csv")
        with open(ruta, "w", encoding="utf8") as archivo:
            archivo.write(
                f"# FASE 9. Parametros del apartado 6.7. {REGLA}, N={N} s={S} K={K} T={T} "
                f"cadenas={N_CADENAS}\n"
                "# Una fila por caso, precio, burn-in y codificacion. Promedios sobre las cadenas; "
                "ajustes sobre las curvas promedio.\n"
                f"# exponente_cola: estimador de Hill con el {FRACCION_COLA * 100:g} % mas grande de "
                "|r - media|. exponente_acf: -pendiente log-log de la autocorrelacion de |r| "
                "(solo retardos con valor positivo).\n"
                "# simbolos: binaria 0 = baja, 1 = sube (cambios nulos omitidos); ternaria 2 = "
                "|r| < umbral*desv. C_l7 y C_l8 no se calculan para la ternaria (nan).\n"
                "caso,m,puntaje,N,gamma,precio,burn_in,r_media,r_desv,curtosis,exponente_cola,"
                "acf_1,acf_10,acf_100,exponente_acf,R2_acf,puntos_acf,f2_media,f2_desv,"
                "codificacion,umbral,longitud_promedio,frac_0,frac_1,frac_2,"
                + ",".join(f"C_l{l}" for l in range(1, L_MAX_BIN + 1))
                + ",alpha_C,beta_C,R2_C,M1,M2,M3,M_max,d_del_maximo,pendiente_M,R2_M,"
                "a_espectro,R2_espectro\n")
            for f in filas:
                archivo.write(",".join(f) + "\n")
        print(f"Guardado: {ruta}")

    print(f"\nListo. Tiempo total: {(time.time() - inicio_reloj) / 60:.1f} min")
```
